Danh sách QI thử nghiệm = danh sách những cột mình nghi ngờ rằng khi kết hợp với nhau có thể giúp nhận diện một người.

```python
#Danh sach QI thu nghiem
candidate_qi=[
    "tuoi",
    "loai_cong_viec",
    "trinh_do_hoc_van",
    "tinh_trang_hon_nhan",
    "nghe_nghiep",
    "quan_he_gia_dinh",
    "chung_toc",
    "gioi_tinh",
    "gio_lam_moi_tuan",
    "quoc_gia"
]
candidate_qi
```

Không đưa thu nhap vao Qi vì là target của Machine Learning và đồng thời là thuộc tính mà ta muốn bảo vệ khỏi suy luận. Nó không nên dùng để tạo nhóm QI ở bước này.

```python
#Gom cac ban ghi co cung bo gia tri Qi thanh nhom
qi_group_sizes=(
    data_clean
    .groupby(candidate_qi,dropna=False)
    .size()
)
qi_group_sizes
```

Nghĩa là từ 32,561 bản ghi, tổ hợp 10 QI của bạn tạo ra 27,515 nhóm khác nhau.

```python
#Tim k nho nhat de biet muc do te nhat cau dataset hien tai khi xet theo boi QI
k_min=qi_group_sizes.min()
print("k nho nhat: ",k_min)
```

Một tổ hợp:
tuổi + loại công việc + học vấn + ... + quốc gia

chỉ xuất hiện ở đúng 1 người/bản ghi.

```python
#Dem so nhom QI co dung 1 ban ghi
unique_qi_groups=(qi_group_sizes==1).sum()
print("so nhom qi co dung mot ban ghi: ",unique_qi_groups)
```

```python
#Tinh ti le phan tram ban ghi dung mot minh
unique_qi_rate=(
    unique_qi_groups/len(data_clean)

)*100
print("Ti le ban ghi bi co lap: ",
      round(unique_qi_rate,2),"%")
```

Với 10 QI có khoảng 76.17% bản ghi có tổ hợp QI chỉ xuất hiện đúng một lần trong dataset.

```python
#Kiem tra QI neu k=5
k_required=5
violation_records=qi_group_sizes[
#Lay cac ban ghi cos so luong nho hon 5
    qi_group_sizes<k_required
].sum()
print("so ban ghi chua dat k=5 la:",violation_records)
```

Nghĩa là trong tổng 32,561 bản ghi, có 30,633 bản ghi đang nằm trong các nhóm QI có kích thước nhỏ hơn 5.

```python
#Tinh ti le phan tram chua datr k=5
violation_rate=(
    violation_records/len(data_clean)
)*100
print("Ty le ban ghi chua dat k=5: ",round(violation_rate,2),"%")
```

Với tỉ lệ 94.08% vì vậy với 10QI đã chonj chưa đạt 5-anonymity

```python
education_order = (
    data_anonymized[
        ["trinh_do_hoc_van", "so_nam_hoc"]
    ]
    .drop_duplicates()
    .sort_values("so_nam_hoc")
)

education_order
```

```python
def generalize_education(education_num):
    if education_num <= 8:
        return "Chua_tot_nghiep_THPT"
    elif education_num == 9:
        return "Tot_nghiep_THPT"
    elif education_num <= 12:
        return "Sau_THPT_Cao_dang"
    elif education_num == 13:
        return "Cu_nhan"
    else:
        return "Sau_dai_hoc"


data_anonymized["hoc_van_nhom"] = (
    data_anonymized["so_nam_hoc"]
    .apply(generalize_education)
)

data_anonymized[
    ["trinh_do_hoc_van", "so_nam_hoc", "hoc_van_nhom"]
].drop_duplicates().sort_values("so_nam_hoc")
```

```python
candidate_qi_age10_hours10_edu = candidate_qi_age10_hours10.copy()

candidate_qi_age10_hours10_edu[
    candidate_qi_age10_hours10_edu.index("trinh_do_hoc_van")
] = "hoc_van_nhom"

candidate_qi_age10_hours10_edu
```

Tức là hiện tại ta đã generalization 3 thuộc tính:

tuoi → tuoi_nhom_10

gio_lam_moi_tuan → gio_lam_nhom_10

trinh_do_hoc_van → hoc_van_nhom

Xem cac gia tri cua tinh trang hon n

```python
marital_counts = data_anonymized[
    "tinh_trang_hon_nhan"
].value_counts(dropna=False)

marital_counts
```

```python
def generalize_marital(status):
    if status == "Chua_ket_hon":
        return "Chua_ket_hon"

    elif status in [
        "Ket_hon_dan_su",
        "Ket_hon_vo_chong_vang_mat",
        "Ket_hon_voi_quan_nhan"
    ]:
        return "Da_ket_hon"

    else:
        return "Khong_con_hon_nhan"


data_anonymized["hon_nhan_nhom"] = (
    data_anonymized["tinh_trang_hon_nhan"]
    .apply(generalize_marital)
)

data_anonymized[
    ["tinh_trang_hon_nhan", "hon_nhan_nhom"]
].drop_duplicates()
```

```python
candidate_qi_age10_hours10_edu_marital = (
    candidate_qi_age10_hours10_edu.copy()
)

candidate_qi_age10_hours10_edu_marital[
    candidate_qi_age10_hours10_edu_marital.index(
        "tinh_trang_hon_nhan"
    )
] = "hon_nhan_nhom"

candidate_qi_age10_hours10_edu_marital
```

```python
qi_group_sizes_age10_hours10_edu_marital = (
    data_anonymized
    .groupby(candidate_qi_age10_hours10_edu_marital, dropna=False)
    .size()
)

qi_group_sizes_age10_hours10_edu_marital
```

```python
k_min_age10_hours10_edu_marital = (
    qi_group_sizes_age10_hours10_edu_marital.min()
)

print(
    "k nhỏ nhất sau khi gom tuổi, giờ làm, học vấn và hôn nhân:",
    k_min_age10_hours10_edu_marital
)
```

Gom cum nghe nghiep

```python
occupation_counts = data_anonymized[
    "nghe_nghiep"
].value_counts(dropna=False)

occupation_counts
```

```python
occupation_map = {
    "Chuyen_mon_dac_thu": "Chuyen_mon_ky_thuat",
    "Ho_tro_ky_thuat": "Chuyen_mon_ky_thuat",

    "Quan_ly_dieu_hanh": "Van_phong_kinh_doanh",
    "Nhan_vien_van_phong": "Van_phong_kinh_doanh",
    "Ban_hang": "Van_phong_kinh_doanh",

    "Tho_thu_cong_sua_chua": "Lao_dong_ky_thuat",
    "Van_hanh_may_moc": "Lao_dong_ky_thuat",
    "Van_tai": "Lao_dong_ky_thuat",
    "Boc_xep_don_dep": "Lao_dong_ky_thuat",

    "Dich_vu_khac": "Dich_vu",
    "Dich_vu_bao_ve": "Dich_vu",
    "Giup_viec_gia_dinh": "Dich_vu",

    "Nong_nghiep_danh_bat_ca": "Nong_nghiep",

    "Quan_doi": "Quan_doi"
}

data_anonymized["nghe_nghiep_nhom"] = (
    data_anonymized["nghe_nghiep"]
    .map(occupation_map)
    .fillna("Khong_ro")
)

data_anonymized[
    ["nghe_nghiep", "nghe_nghiep_nhom"]
].drop_duplicates().sort_values("nghe_nghiep_nhom")
```

```python
# Tạo bảng tổng hợp nhóm có tỷ lệ vi phạm cao nhất của từng QI
risk_summary = pd.DataFrame({
    "QI": [
        "tuoi_nhom_20",
        "loai_cong_viec_nhom",
        "hoc_van_nhom_4",
        "hon_nhan_nhom_2",
        "nghe_nghiep_nhom_4",
        "gio_lam_nhom_20",
        "quoc_gia_nhom_3"
    ],

    # Lấy tên nhóm có tỷ lệ vi phạm cao nhất trong từng QI
    "nhom_rui_ro_cao_nhat": [
        age_risk["ty_le_vi_pham_%"].idxmax(),
        workclass_risk["ty_le_vi_pham_%"].idxmax(),
        education_risk["ty_le_vi_pham_%"].idxmax(),
        marital_risk["ty_le_vi_pham_%"].idxmax(),
        occupation_risk["ty_le_vi_pham_%"].idxmax(),
        hours_risk["ty_le_vi_pham_%"].idxmax(),
        country_risk["ty_le_vi_pham_%"].idxmax()
    ],

    # Lấy tỷ lệ vi phạm cao nhất tương ứng của từng QI
    "ty_le_vi_pham_cao_nhat_%": [
        age_risk["ty_le_vi_pham_%"].max(),
        workclass_risk["ty_le_vi_pham_%"].max(),
        education_risk["ty_le_vi_pham_%"].max(),
        marital_risk["ty_le_vi_pham_%"].max(),
        occupation_risk["ty_le_vi_pham_%"].max(),
        hours_risk["ty_le_vi_pham_%"].max(),
        country_risk["ty_le_vi_pham_%"].max()
    ]
})

# Sắp xếp QI có nhóm rủi ro cao nhất lên đầu
risk_summary = risk_summary.sort_values(
    "ty_le_vi_pham_cao_nhat_%",
    ascending=False
)

# Hiển thị bảng tổng hợp
risk_summary
```

```python
# Thêm tổng số bản ghi của nhóm rủi ro cao nhất trong từng QI
risk_summary["tong_ban_ghi_nhom_rui_ro"] = [
    age_risk.loc[
        age_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    workclass_risk.loc[
        workclass_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    education_risk.loc[
        education_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    marital_risk.loc[
        marital_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    occupation_risk.loc[
        occupation_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    hours_risk.loc[
        hours_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ],
    country_risk.loc[
        country_risk["ty_le_vi_pham_%"].idxmax(),
        "tong_ban_ghi"
    ]
]


# Thêm số bản ghi vi phạm k=5 của nhóm rủi ro cao nhất trong từng QI
risk_summary["so_ban_ghi_vi_pham"] = [
    age_risk.loc[
        age_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    workclass_risk.loc[
        workclass_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    education_risk.loc[
        education_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    marital_risk.loc[
        marital_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    occupation_risk.loc[
        occupation_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    hours_risk.loc[
        hours_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ],
    country_risk.loc[
        country_risk["ty_le_vi_pham_%"].idxmax(),
        "ban_ghi_vi_pham_k5"
    ]
]


# Hiển thị lại bảng tổng hợp sau khi bổ sung quy mô nhóm rủi ro
risk_summary
```

```python
# Gom các bảng phân tích rủi ro vào một nơi để dễ xử lý
risk_tables = {
    "tuoi_nhom_20": age_risk,
    "loai_cong_viec_nhom": workclass_risk,
    "hoc_van_nhom_4": education_risk,
    "hon_nhan_nhom_2": marital_risk,
    "nghe_nghiep_nhom_4": occupation_risk,
    "gio_lam_nhom_20": hours_risk,
    "quoc_gia_nhom_3": country_risk
}


# Tạo danh sách để lưu kết quả tổng hợp của từng QI
risk_summary_rows = []


# Lần lượt phân tích từng QI
for qi_name, risk_table in risk_tables.items():

    # Tìm tên nhóm có tỷ lệ vi phạm k=5 cao nhất
    highest_risk_group = risk_table["ty_le_vi_pham_%"].idxmax()

    # Lấy đúng thông tin của nhóm rủi ro cao nhất và lưu vào danh sách
    risk_summary_rows.append({
        "QI": qi_name,
        "nhom_rui_ro_cao_nhat": highest_risk_group,
        "ty_le_vi_pham_cao_nhat_%": risk_table.loc[
            highest_risk_group,
            "ty_le_vi_pham_%"
        ],
        "tong_ban_ghi_nhom_rui_ro": int(
            risk_table.loc[
                highest_risk_group,
                "tong_ban_ghi"
            ]
        ),
        "so_ban_ghi_vi_pham": int(
            risk_table.loc[
                highest_risk_group,
                "ban_ghi_vi_pham_k5"
            ]
        )
    })


# Chuyển danh sách kết quả thành DataFrame
risk_summary = pd.DataFrame(risk_summary_rows)


# Sắp xếp QI có nhóm rủi ro cao nhất lên đầu
risk_summary = risk_summary.sort_values(
    "ty_le_vi_pham_cao_nhat_%",
    ascending=False
).reset_index(drop=True)


# Hiển thị bảng tổng hợp đã sửa đúng
risk_summary
```

```python
# Lấy tổng số bản ghi đang vi phạm k=5
tong_vi_pham_k5 = len(violating_records_country3)


# Tính tỷ lệ mỗi nhóm rủi ro cao đóng góp vào tổng số bản ghi vi phạm k=5
risk_summary["ty_le_dong_gop_vi_pham_%"] = (
    risk_summary["so_ban_ghi_vi_pham"]
    / tong_vi_pham_k5
) * 100


# Làm tròn tỷ lệ đến 2 chữ số thập phân để bảng dễ đọc
risk_summary["ty_le_dong_gop_vi_pham_%"] = (
    risk_summary["ty_le_dong_gop_vi_pham_%"]
    .round(2)
)


# Hiển thị lại bảng tổng hợp sau khi thêm tỷ lệ đóng góp vi phạm
risk_summary
```

```python
# Tính tỷ lệ vi phạm k=5 chung của toàn bộ dataset
global_violation_rate = (
    len(violating_records_country3)
    / len(data_anonymized)
) * 100


# Tính xem nhóm rủi ro cao nhất của mỗi QI
# có tỷ lệ vi phạm cao gấp bao nhiêu lần mức trung bình toàn dataset
risk_summary["he_so_rui_ro_so_voi_toan_bo"] = (
    risk_summary["ty_le_vi_pham_cao_nhat_%"]
    / global_violation_rate
)


# Làm tròn hệ số rủi ro đến 2 chữ số thập phân để dễ đọc
risk_summary["he_so_rui_ro_so_voi_toan_bo"] = (
    risk_summary["he_so_rui_ro_so_voi_toan_bo"]
    .round(2)
)


# Hiển thị bảng tổng hợp sau khi bổ sung hệ số rủi ro
risk_summary
```

```python
# Tạo danh sách để lưu bảng rủi ro chi tiết của tất cả QI
all_risk_tables = []


# Gom tên QI và bảng phân tích rủi ro tương ứng
risk_tables = {
    "tuoi_nhom_20": age_risk,
    "loai_cong_viec_nhom": workclass_risk,
    "hoc_van_nhom_4": education_risk,
    "hon_nhan_nhom_2": marital_risk,
    "nghe_nghiep_nhom_4": occupation_risk,
    "gio_lam_nhom_20": hours_risk,
    "quoc_gia_nhom_3": country_risk
}


# Lần lượt xử lý từng bảng rủi ro
for qi_name, risk_table in risk_tables.items():

    # Đưa tên nhóm đang nằm ở index thành một cột bình thường
    temp = risk_table.reset_index()

    # Đổi tên cột đầu tiên thành tên dễ hiểu hơn
    temp = temp.rename(
        columns={temp.columns[0]: "nhom_gia_tri"}
    )

    # Gắn tên QI tương ứng cho từng dòng
    temp["QI"] = qi_name

    # Tính nhóm này có rủi ro cao gấp bao nhiêu lần mức chung của dataset
    temp["he_so_rui_ro"] = (
        temp["ty_le_vi_pham_%"]
        / global_violation_rate
    )

    # Lưu bảng vừa xử lý vào danh sách
    all_risk_tables.append(temp)


# Ghép tất cả bảng QI thành một bảng rủi ro chung
risk_detail_all = pd.concat(
    all_risk_tables,
    ignore_index=True
)


# Làm tròn hệ số rủi ro để bảng dễ đọc hơn
risk_detail_all["he_so_rui_ro"] = (
    risk_detail_all["he_so_rui_ro"]
    .round(2)
)


# Sắp xếp nhóm có hệ số rủi ro cao nhất lên đầu
risk_detail_all = risk_detail_all.sort_values(
    "he_so_rui_ro",
    ascending=False
)


# Hiển thị toàn bộ bảng rủi ro chi tiết
risk_detail_all
```

```python
# Tính tỷ lệ vi phạm k=5 chung của toàn bộ dataset dưới dạng số thập phân
global_violation_rate_decimal = (
    len(violating_records_country3)
    / len(data_anonymized)
)


# Tính số bản ghi vi phạm dự kiến nếu mỗi nhóm chỉ có
# mức rủi ro bằng đúng mức trung bình của toàn bộ dataset
risk_detail_all["so_vi_pham_ky_vong"] = (
    risk_detail_all["tong_ban_ghi"]
    * global_violation_rate_decimal
)


# Tính số bản ghi vi phạm vượt quá mức dự kiến của toàn dataset
risk_detail_all["vi_pham_du_thua"] = (
    risk_detail_all["ban_ghi_vi_pham_k5"]
    - risk_detail_all["so_vi_pham_ky_vong"]
)


# Làm tròn các giá trị để bảng dễ đọc hơn
risk_detail_all["so_vi_pham_ky_vong"] = (
    risk_detail_all["so_vi_pham_ky_vong"].round(2)
)

risk_detail_all["vi_pham_du_thua"] = (
    risk_detail_all["vi_pham_du_thua"].round(2)
)


# Sắp xếp để xem những nhóm có số vi phạm dư thừa lớn nhất
risk_detail_all.sort_values(
    "vi_pham_du_thua",
    ascending=False
).head(15)
```

```python
# Chỉ giữ phần vi phạm dư thừa lớn hơn 0
# Nếu một nhóm có rủi ro thấp hơn mức trung bình thì không tính vào phần rủi ro dư thừa
risk_detail_all["vi_pham_du_thua_duong"] = (
    risk_detail_all["vi_pham_du_thua"]
    .clip(lower=0)
)


# Cộng tổng số vi phạm dư thừa của các nhóm thuộc cùng một QI
qi_risk_priority = (
    risk_detail_all
    .groupby("QI")["vi_pham_du_thua_duong"]
    .sum()
    .reset_index()
)


# Đổi tên cột để ý nghĩa dễ hiểu hơn
qi_risk_priority = qi_risk_priority.rename(
    columns={
        "vi_pham_du_thua_duong": "tong_vi_pham_du_thua"
    }
)


# Làm tròn kết quả đến 2 chữ số thập phân
qi_risk_priority["tong_vi_pham_du_thua"] = (
    qi_risk_priority["tong_vi_pham_du_thua"]
    .round(2)
)


# Sắp xếp QI có mức rủi ro tập trung lớn nhất lên đầu
qi_risk_priority = qi_risk_priority.sort_values(
    "tong_vi_pham_du_thua",
    ascending=False
)


# Hiển thị bảng xếp hạng QI cần ưu tiên phân tích
qi_risk_priority
```

```python
# Tạo lại bảng so sánh privacy của 3 strategy đã thử nghiệm
strategy_privacy_comparison = pd.DataFrame({

    # Đặt tên cho từng strategy để dễ phân biệt
    "strategy": [
        "Country3_Hours20",
        "Country3_Hours4",
        "Country3_Hours4_AgeHigh"
    ],

    # Số equivalence class của từng strategy
    "so_equivalence_class": [
        len(qi_group_sizes_country3),
        len(qi_group_sizes_hours4),
        len(qi_group_sizes_hours4_agehigh)
    ],

    # Giá trị k nhỏ nhất của từng strategy
    "k_min": [
        qi_group_sizes_country3.min(),
        qi_group_sizes_hours4.min(),
        qi_group_sizes_hours4_agehigh.min()
    ],

    # Số nhóm chỉ có đúng 1 bản ghi
    "so_nhom_co_lap": [
        unique_groups_country3,
        unique_groups_hours4,
        unique_groups_hours4_agehigh
    ],

    # Tỷ lệ phần trăm bản ghi bị cô lập
    "ty_le_co_lap_%": [
        unique_rate_country3,
        unique_rate_hours4,
        unique_rate_hours4_agehigh
    ],

    # Số bản ghi đang nằm trong equivalence class có kích thước nhỏ hơn 5
    "so_ban_ghi_vi_pham_k5": [
        violation_records_country3,
        violation_records_hours4,
        violation_records_hours4_agehigh
    ],

    # Tỷ lệ phần trăm bản ghi chưa đạt k=5
    "ty_le_vi_pham_k5_%": [
        violation_rate_country3,
        violation_rate_hours4,
        violation_rate_hours4_agehigh
    ]
})


# Làm tròn các cột phần trăm đến 2 chữ số thập phân để dễ đọc
strategy_privacy_comparison[
    ["ty_le_co_lap_%", "ty_le_vi_pham_k5_%"]
] = (
    strategy_privacy_comparison[
        ["ty_le_co_lap_%", "ty_le_vi_pham_k5_%"]
    ].round(2)
)


# Hiển thị bảng so sánh 3 strategy
strategy_privacy_comparison
```

```python
# Lấy số bản ghi vi phạm k=5 của strategy gốc làm mốc so sánh
baseline_violation_records = strategy_privacy_comparison.loc[
    0,
    "so_ban_ghi_vi_pham_k5"
]


# Lấy số nhóm cô lập của strategy gốc làm mốc so sánh
baseline_unique_groups = strategy_privacy_comparison.loc[
    0,
    "so_nhom_co_lap"
]


# Tính mỗi strategy đã giảm được bao nhiêu bản ghi vi phạm k=5 so với strategy gốc
strategy_privacy_comparison["giam_ban_ghi_vi_pham_k5"] = (
    baseline_violation_records
    - strategy_privacy_comparison["so_ban_ghi_vi_pham_k5"]
)


# Tính mỗi strategy đã giảm được bao nhiêu nhóm cô lập so với strategy gốc
strategy_privacy_comparison["giam_nhom_co_lap"] = (
    baseline_unique_groups
    - strategy_privacy_comparison["so_nhom_co_lap"]
)


# Hiển thị lại bảng so sánh sau khi thêm mức cải thiện
strategy_privacy_comparison
```

```python
# Tạo một nơi lưu các strategy ẩn danh đã xây dựng
# Mỗi strategy gồm tên phương án và danh sách QI tương ứng
strategies = {
    "Country3_Hours20": candidate_qi_country3,
    "Country3_Hours4": candidate_qi_hours4,
    "Country3_Hours4_AgeHigh": candidate_qi_hours4_agehigh
}


# Hiển thị từng strategy và các QI mà strategy đó đang sử dụng
for strategy_name, qi_columns in strategies.items():

    # In tên strategy
    print("\nStrategy:", strategy_name)

    # In danh sách QI của strategy tương ứng
    print("QI:", qi_columns)
```

Country3_Hours20
→ phương án gốc hiện tại

Country3_Hours4
→ gộp vùng giờ làm cao

Country3_Hours4_AgeHigh
→ gộp vùng giờ làm cao + vùng tuổi cao

```python
# Tạo hàm dùng chung để đánh giá privacy cho một strategy bất kỳ
def evaluate_privacy(dataframe, qi_columns, k_required=5):

    # Gom các bản ghi có cùng bộ QI vào cùng một equivalence class
    group_sizes = (
        dataframe
        .groupby(
            qi_columns,
            dropna=False
        )
        .size()
    )

    # Đếm tổng số equivalence class được tạo ra
    so_equivalence_class = len(group_sizes)

    # Tìm kích thước nhỏ nhất trong tất cả equivalence class
    k_min = int(group_sizes.min())

    # Đếm số equivalence class chỉ có đúng 1 bản ghi
    so_nhom_co_lap = int(
        (group_sizes == 1).sum()
    )

    # Tính tỷ lệ phần trăm bản ghi bị cô lập
    ty_le_co_lap = (
        so_nhom_co_lap
        / len(dataframe)
    ) * 100

    # Cộng số bản ghi nằm trong các nhóm có kích thước nhỏ hơn k yêu cầu
    so_ban_ghi_vi_pham_k = int(
        group_sizes[
            group_sizes < k_required
        ].sum()
    )

    # Tính tỷ lệ phần trăm bản ghi chưa đạt k-anonymity yêu cầu
    ty_le_vi_pham_k = (
        so_ban_ghi_vi_pham_k
        / len(dataframe)
    ) * 100

    # Trả về toàn bộ kết quả privacy của strategy
    return {
        "so_equivalence_class": so_equivalence_class,
        "k_min": k_min,
        "so_nhom_co_lap": so_nhom_co_lap,
        "ty_le_co_lap_%": round(ty_le_co_lap, 2),
        "so_ban_ghi_vi_pham_k5": so_ban_ghi_vi_pham_k,
        "ty_le_vi_pham_k5_%": round(ty_le_vi_pham_k, 2)
    }
```

Hàm evaluate_privacy() có tác dụng:

Đánh giá mức độ riêng tư của một phương án ẩn danh.

```python
# Dùng hàm evaluate_privacy để đánh giá strategy Country3_Hours20
privacy_test_country3 = evaluate_privacy(
    data_anonymized,
    candidate_qi_country3,
    k_required=5
)


# Hiển thị kết quả privacy mà hàm tự tính
privacy_test_country3
```

```python
# Tạo danh sách để lưu kết quả đánh giá privacy của từng strategy
privacy_results = []


# Lần lượt lấy tên strategy và bộ QI tương ứng
for strategy_name, qi_columns in strategies.items():

    # Dùng hàm evaluate_privacy để tự động tính các chỉ số privacy
    result = evaluate_privacy(
        data_anonymized,
        qi_columns,
        k_required=5
    )

    # Gắn tên strategy vào kết quả để biết kết quả thuộc phương án nào
    result["strategy"] = strategy_name

    # Lưu kết quả của strategy hiện tại vào danh sách
    privacy_results.append(result)


# Chuyển danh sách kết quả thành DataFrame để dễ xem và so sánh
strategy_evaluation = pd.DataFrame(
    privacy_results
)


# Sắp xếp lại thứ tự các cột để tên strategy nằm ở đầu bảng
strategy_evaluation = strategy_evaluation[
    [
        "strategy",
        "so_equivalence_class",
        "k_min",
        "so_nhom_co_lap",
        "ty_le_co_lap_%",
        "so_ban_ghi_vi_pham_k5",
        "ty_le_vi_pham_k5_%"
    ]
]


# Hiển thị bảng kết quả đánh giá tự động của tất cả strategy
strategy_evaluation
```

```python
# Sắp xếp các strategy theo tỷ lệ vi phạm k=5 từ thấp đến cao
# Nếu hai strategy có tỷ lệ vi phạm giống nhau thì ưu tiên strategy có tỷ lệ cô lập thấp hơn
privacy_ranking = strategy_evaluation.sort_values(
    by=[
        "ty_le_vi_pham_k5_%",
        "ty_le_co_lap_%"
    ],
    ascending=[
        True,
        True
    ]
)


# Lấy strategy đứng đầu sau khi xếp hạng privacy
best_privacy_strategy = privacy_ranking.iloc[0]


# Hiển thị tên strategy đang tốt nhất về mặt privacy
print(
    "Strategy tốt nhất về Privacy hiện tại:",
    best_privacy_strategy["strategy"]
)


# Hiển thị toàn bộ thông tin privacy của strategy tốt nhất
best_privacy_strategy
```

Kiểm tra k nằm trong khoảng 1,2,3,4

```python
# Lấy riêng các equivalence class có kích thước nhỏ hơn k=5
violation_groups_best = (
    qi_group_sizes_hours4_agehigh[
        qi_group_sizes_hours4_agehigh < k_required
    ]
)

# Đếm xem có bao nhiêu nhóm có kích thước 1, 2, 3 hoặc 4
violation_distribution_best = (
    violation_groups_best
    .value_counts()
    .sort_index()
)

# Hiển thị phân bố các nhóm đang vi phạm k=5
violation_distribution_best
```

```python
# Tính kích thước equivalence class tương ứng cho từng bản ghi
group_size_per_record_best = (
    data_anonymized
    .groupby(
        candidate_qi_hours4_agehigh,
        dropna=False
    )["tuoi"]
    .transform("size")
)


# Đánh dấu những bản ghi đang nằm trong nhóm có kích thước nhỏ hơn k=5
violation_mask_best = (
    group_size_per_record_best < k_required
)


# Lấy riêng các bản ghi chưa đạt k=5
violating_records_best = (
    data_anonymized
    .loc[violation_mask_best]
    .copy()
)


# Gắn kích thước equivalence class vào từng bản ghi vi phạm
# Ví dụ group_size = 1, 2, 3 hoặc 4
violating_records_best["group_size"] = (
    group_size_per_record_best[
        violation_mask_best
    ]
)


# Kiểm tra số lượng bản ghi vi phạm
print(
    "Số bản ghi chưa đạt k=5:",
    len(violating_records_best)
)


# Hiển thị 10 bản ghi vi phạm đầu tiên để kiểm tra
violating_records_best[
    candidate_qi_hours4_agehigh + ["group_size"]
].head(10)
```

```python
# Tạo danh sách để lưu kết quả phân tích rủi ro của từng QI
risk_summary_best = []


# Lần lượt phân tích từng QI trong strategy tốt nhất hiện tại
for qi_column in candidate_qi_hours4_agehigh:

    # Đếm tổng số bản ghi của từng nhóm giá trị QI trong toàn bộ dữ liệu
    total_by_group = (
        data_anonymized
        .groupby(qi_column, dropna=False)
        .size()
        .rename("tong_ban_ghi")
    )

    # Đếm số bản ghi vi phạm k=5 của từng nhóm giá trị QI
    violation_by_group = (
        violating_records_best
        .groupby(qi_column, dropna=False)
        .size()
        .rename("so_ban_ghi_vi_pham")
    )

    # Ghép tổng số bản ghi và số bản ghi vi phạm vào cùng một bảng
    risk_table = (
        pd.concat(
            [total_by_group, violation_by_group],
            axis=1
        )
        .fillna(0)
    )

    # Tính tỷ lệ phần trăm vi phạm k=5 trong từng nhóm giá trị
    risk_table["ty_le_vi_pham_%"] = (
        risk_table["so_ban_ghi_vi_pham"]
        / risk_table["tong_ban_ghi"]
        * 100
    )

    # Tìm nhóm giá trị có tỷ lệ vi phạm cao nhất của QI hiện tại
    highest_risk_group = (
        risk_table
        .sort_values(
            "ty_le_vi_pham_%",
            ascending=False
        )
        .iloc[0]
    )

    # Lưu thông tin nhóm rủi ro cao nhất của QI hiện tại
    risk_summary_best.append(
        {
            "QI": qi_column,
            "nhom_rui_ro_cao_nhat": highest_risk_group.name,
            "tong_ban_ghi_nhom": int(highest_risk_group["tong_ban_ghi"]),
            "so_ban_ghi_vi_pham": int(highest_risk_group["so_ban_ghi_vi_pham"]),
            "ty_le_vi_pham_%": round(highest_risk_group["ty_le_vi_pham_%"], 2)
        }
    )


# Chuyển kết quả thành DataFrame để dễ so sánh
risk_summary_best = pd.DataFrame(
    risk_summary_best
)


# Sắp xếp QI từ nhóm có tỷ lệ vi phạm cao nhất xuống thấp nhất
risk_summary_best = (
    risk_summary_best
    .sort_values(
        "ty_le_vi_pham_%",
        ascending=False
    )
    .reset_index(drop=True)
)


# Hiển thị bảng tổng hợp rủi ro của 7 QI
risk_summary_best
```

```python
# Tạo một bản sao dữ liệu mới cho strategy có suppression
# Chỉ giữ các bản ghi đang thuộc equivalence class có kích thước từ 5 trở lên
data_country2_suppressed = (
    data_anonymized
    .loc[~violation_mask_country2]
    .copy()
)


# Tính số bản ghi còn lại sau khi suppression
so_ban_ghi_con_lai = len(data_country2_suppressed)


# Tính số bản ghi đã bị suppression
so_ban_ghi_suppression = (
    len(data_anonymized) - so_ban_ghi_con_lai
)


# Tính tỷ lệ phần trăm dữ liệu bị suppression
ty_le_suppression = (
    so_ban_ghi_suppression
    / len(data_anonymized)
    * 100
)


# Hiển thị kết quả để kiểm tra
print("Số bản ghi ban đầu:", len(data_anonymized))
print("Số bản ghi bị suppression:", so_ban_ghi_suppression)
print("Số bản ghi còn lại:", so_ban_ghi_con_lai)
print("Tỷ lệ suppression:", round(ty_le_suppression, 2), "%")
```

```python
# Đánh giá lại privacy trên dữ liệu sau khi đã suppression
privacy_after_suppression = evaluate_privacy(
    data_country2_suppressed,
    candidate_qi_country2_hours4_agehigh,
    k_required=5
)


# Hiển thị kết quả để kiểm tra dataset sau suppression có thật sự đạt k=5 hay chưa
privacy_after_suppression
```

```python
# Lấy giá trị k nhỏ nhất của dữ liệu sau khi suppression
k_min_after_suppression = privacy_after_suppression["k_min"]

# Hiển thị giá trị k nhỏ nhất
print(
    "k nhỏ nhất sau suppression:",
    k_min_after_suppression
)
```

```python
# Đặt ngưỡng l-diversity cần kiểm tra
# Vì thu_nhap chỉ có 2 giá trị nên ta thử nghiệm với l = 2
l_required = 2


# Với mỗi equivalence class, đếm số giá trị thu_nhap khác nhau
l_values_country2 = (
    data_country2_suppressed
    .groupby(
        candidate_qi_country2_hours4_agehigh,
        dropna=False
    )["thu_nhap"]
    .nunique(dropna=False)
)


# Tìm giá trị l nhỏ nhất của toàn bộ các equivalence class
l_min_country2 = int(
    l_values_country2.min()
)


# Đếm số equivalence class chưa đạt l = 2
so_nhom_vi_pham_l2 = int(
    (l_values_country2 < l_required).sum()
)


# Xem có bao nhiêu nhóm đạt l = 1 hoặc l = 2
l_distribution_country2 = (
    l_values_country2
    .value_counts()
    .sort_index()
)


# Hiển thị kết quả kiểm tra l-diversity
print("l_min:", l_min_country2)
print("Số equivalence class chưa đạt l=2:", so_nhom_vi_pham_l2)

print("\nPhân bố l-diversity:")
print(l_distribution_country2)
```

```python
# Tính số bản ghi của từng equivalence class sau suppression
group_sizes_after_suppression = (
    data_country2_suppressed
    .groupby(
        candidate_qi_country2_hours4_agehigh,
        dropna=False
    )
    .size()
)


# Ghép kích thước nhóm và giá trị l-diversity của từng equivalence class
l_group_analysis = pd.DataFrame({
    "group_size": group_sizes_after_suppression,
    "l_value": l_values_country2
})


# Chỉ lấy những equivalence class chưa đạt l=2
l2_violating_groups = (
    l_group_analysis[
        l_group_analysis["l_value"] < l_required
    ]
)


# Tính tổng số bản ghi nằm trong các nhóm chưa đạt l=2
so_ban_ghi_vi_pham_l2 = int(
    l2_violating_groups["group_size"].sum()
)


# Tính tỷ lệ bản ghi bị ảnh hưởng bởi vi phạm l=2
ty_le_ban_ghi_vi_pham_l2 = (
    so_ban_ghi_vi_pham_l2
    / len(data_country2_suppressed)
    * 100
)


# Hiển thị kết quả
print("Số equivalence class chưa đạt l=2:", len(l2_violating_groups))
print("Số bản ghi nằm trong các nhóm chưa đạt l=2:", so_ban_ghi_vi_pham_l2)
print(
    "Tỷ lệ bản ghi nằm trong các nhóm chưa đạt l=2:",
    round(ty_le_ban_ghi_vi_pham_l2, 2),
    "%"
)
```

```python
# Tạo bảng thông tin cho từng equivalence class sau khi đã đạt k=5
l_group_detail = (
    data_country2_suppressed
    .groupby(
        candidate_qi_country2_hours4_agehigh,
        dropna=False
    )
    .agg(
        # Đếm số bản ghi trong từng equivalence class
        group_size=("thu_nhap", "size"),

        # Đếm số giá trị thu_nhap khác nhau trong từng equivalence class
        l_value=("thu_nhap", "nunique"),

        # Lấy một giá trị thu_nhap đại diện
        # Với nhóm l=1 thì toàn bộ nhóm chỉ có đúng giá trị này
        thu_nhap_dai_dien=("thu_nhap", "first")
    )
    .reset_index()
)


# Chỉ lấy các equivalence class chưa đạt l=2
l1_groups = (
    l_group_detail[
        l_group_detail["l_value"] == 1
    ]
    .copy()
)


# Tổng hợp xem các nhóm l=1 đang thuộc loại thu_nhap nào
l1_income_summary = (
    l1_groups
    .groupby("thu_nhap_dai_dien")
    .agg(
        # Đếm số equivalence class l=1 của từng loại thu nhập
        so_equivalence_class=("group_size", "size"),

        # Cộng tổng số bản ghi nằm trong các nhóm đó
        so_ban_ghi=("group_size", "sum")
    )
)


# Hiển thị kết quả
l1_income_summary
```

```python
# Tính giá trị l-diversity tương ứng cho từng bản ghi
# Các bản ghi có cùng bộ QI sẽ nhận cùng một giá trị l
l_value_per_record_country2 = (
    data_country2_suppressed
    .groupby(
        candidate_qi_country2_hours4_agehigh,
        dropna=False
    )["thu_nhap"]
    .transform("nunique")
)


# Đánh dấu những bản ghi đang nằm trong equivalence class chưa đạt l=2
l2_violation_mask_country2 = (
    l_value_per_record_country2 < l_required
)


# Lấy riêng các bản ghi chưa đạt l=2
# Tạo bản sao để không làm thay đổi dataset sau suppression
violating_l2_records_country2 = (
    data_country2_suppressed
    .loc[l2_violation_mask_country2]
    .copy()
)


# Gắn giá trị l vào từng bản ghi để tiện phân tích ở các bước sau
violating_l2_records_country2["l_value"] = (
    l_value_per_record_country2[
        l2_violation_mask_country2
    ]
)


# Kiểm tra số lượng bản ghi chưa đạt l=2
print(
    "Số bản ghi chưa đạt l=2:",
    len(violating_l2_records_country2)
)


# Kiểm tra các bản ghi vi phạm đang thuộc loại thu nhập nào
print("\nPhân bố thu nhập trong các bản ghi chưa đạt l=2:")
print(
    violating_l2_records_country2["thu_nhap"]
    .value_counts()
)
```

```python
# Tạo danh sách để lưu kết quả phân tích vi phạm l=2 của từng QI
l2_risk_summary = []


# Lần lượt phân tích từng QI của strategy Country2_Hours4_AgeHigh
for qi_column in candidate_qi_country2_hours4_agehigh:

    # Đếm tổng số bản ghi của từng nhóm giá trị QI
    # trên dataset sau khi đã đạt k=5
    total_by_group = (
        data_country2_suppressed
        .groupby(qi_column, dropna=False)
        .size()
        .rename("tong_ban_ghi")
    )

    # Đếm số bản ghi chưa đạt l=2 trong từng nhóm giá trị QI
    violation_by_group = (
        violating_l2_records_country2
        .groupby(qi_column, dropna=False)
        .size()
        .rename("so_ban_ghi_vi_pham_l2")
    )

    # Ghép tổng số bản ghi và số bản ghi vi phạm vào cùng một bảng
    l2_risk_table = (
        pd.concat(
            [total_by_group, violation_by_group],
            axis=1
        )
        .fillna(0)
    )

    # Tính tỷ lệ bản ghi chưa đạt l=2 của từng nhóm giá trị
    l2_risk_table["ty_le_vi_pham_l2_%"] = (
        l2_risk_table["so_ban_ghi_vi_pham_l2"]
        / l2_risk_table["tong_ban_ghi"]
        * 100
    )

    # Tìm nhóm giá trị có tỷ lệ vi phạm l=2 cao nhất của QI hiện tại
    highest_risk_group = (
        l2_risk_table
        .sort_values(
            "ty_le_vi_pham_l2_%",
            ascending=False
        )
        .iloc[0]
    )

    # Lưu kết quả của QI hiện tại vào danh sách tổng hợp
    l2_risk_summary.append(
        {
            "QI": qi_column,
            "nhom_rui_ro_cao_nhat": highest_risk_group.name,
            "tong_ban_ghi_nhom": int(
                highest_risk_group["tong_ban_ghi"]
            ),
            "so_ban_ghi_vi_pham_l2": int(
                highest_risk_group["so_ban_ghi_vi_pham_l2"]
            ),
            "ty_le_vi_pham_l2_%": round(
                highest_risk_group["ty_le_vi_pham_l2_%"],
                2
            )
        }
    )


# Chuyển kết quả thành DataFrame để dễ xem
l2_risk_summary = pd.DataFrame(
    l2_risk_summary
)


# Sắp xếp từ nhóm có tỷ lệ vi phạm l=2 cao nhất xuống thấp nhất
l2_risk_summary = (
    l2_risk_summary
    .sort_values(
        "ty_le_vi_pham_l2_%",
        ascending=False
    )
    .reset_index(drop=True)
)


# Hiển thị bảng tổng hợp
l2_risk_summary
```

```python
# Tạo quy tắc generalization giờ làm mới từ 4 nhóm xuống còn 3 nhóm
# Gộp nhóm 1-20 và 21-40 thành một nhóm rộng hơn là 1-40
hours_map_3 = {
    "1-20": "1-40",
    "21-40": "1-40",
    "41-60": "41-60",
    "61-100": "61-100"
}


# Tạo cột giờ làm mới trên dữ liệu đầy đủ
# Không ghi đè cột gio_lam_nhom_4 để còn so sánh các strategy
data_anonymized["gio_lam_nhom_3"] = (
    data_anonymized["gio_lam_nhom_4"]
    .map(hours_map_3)
)


# Kiểm tra số lượng bản ghi trong từng nhóm giờ làm mới
data_anonymized["gio_lam_nhom_3"].value_counts(dropna=False)
```

```python
# Tạo bộ QI cho strategy mới
# Giữ nguyên các QI hiện tại và chỉ thay gio_lam_nhom_4 bằng gio_lam_nhom_3
candidate_qi_country2_hours3_agehigh = [
    "tuoi_nhom_high",
    "loai_cong_viec_nhom",
    "hoc_van_nhom_4",
    "hon_nhan_nhom_2",
    "nghe_nghiep_nhom_4",
    "gio_lam_nhom_3",
    "quoc_gia_nhom_2"
]


# Đánh giá privacy của strategy mới trên toàn bộ dữ liệu
# Chưa suppression để xem bản thân generalization mới cải thiện k-anonymity bao nhiêu
privacy_country2_hours3_agehigh = evaluate_privacy(
    data_anonymized,
    candidate_qi_country2_hours3_agehigh,
    k_required=5
)


# Hiển thị kết quả đánh giá k-anonymity của strategy mới
privacy_country2_hours3_agehigh
```

```python
# Tính kích thước equivalence class tương ứng cho từng bản ghi
# theo strategy Country2_Hours3_AgeHigh
group_size_per_record_hours3 = (
    data_anonymized
    .groupby(
        candidate_qi_country2_hours3_agehigh,
        dropna=False
    )["tuoi"]
    .transform("size")
)


# Đánh dấu các bản ghi đang nằm trong nhóm có kích thước nhỏ hơn k=5
violation_mask_hours3 = (
    group_size_per_record_hours3 < k_required
)


# Tạo dataset mới và loại các bản ghi vẫn chưa đạt k=5
# Không sửa trực tiếp data_anonymized để còn dùng so sánh strategy
data_country2_hours3_suppressed = (
    data_anonymized
    .loc[~violation_mask_hours3]
    .copy()
)


# Tính số bản ghi bị suppression ở strategy mới
so_suppression_hours3 = (
    len(data_anonymized)
    - len(data_country2_hours3_suppressed)
)


# Hiển thị kết quả để kiểm tra
print("Số bản ghi ban đầu:", len(data_anonymized))
print("Số bản ghi bị suppression:", so_suppression_hours3)
print(
    "Số bản ghi còn lại:",
    len(data_country2_hours3_suppressed)
)
print(
    "Tỷ lệ suppression:",
    round(
        so_suppression_hours3 / len(data_anonymized) * 100,
        2
    ),
    "%"
)
```

```python
# Tính số bản ghi của từng equivalence class
# trong strategy Hours3 sau khi đã suppression
group_sizes_hours3_after_suppression = (
    data_country2_hours3_suppressed
    .groupby(
        candidate_qi_country2_hours3_agehigh,
        dropna=False
    )
    .size()
)


# Tạo bảng chứa kích thước nhóm và giá trị l-diversity
# của từng equivalence class
l_group_analysis_hours3 = pd.DataFrame({
    "group_size": group_sizes_hours3_after_suppression,
    "l_value": l_values_hours3
})


# Chỉ lấy những equivalence class chưa đạt l=2
l2_violating_groups_hours3 = (
    l_group_analysis_hours3[
        l_group_analysis_hours3["l_value"] < l_required
    ]
)


# Cộng số bản ghi nằm trong tất cả các nhóm chưa đạt l=2
so_ban_ghi_vi_pham_l2_hours3 = int(
    l2_violating_groups_hours3["group_size"].sum()
)


# Tính tỷ lệ bản ghi nằm trong các nhóm chưa đạt l=2
ty_le_ban_ghi_vi_pham_l2_hours3 = (
    so_ban_ghi_vi_pham_l2_hours3
    / len(data_country2_hours3_suppressed)
    * 100
)


# Hiển thị kết quả để so sánh với strategy Hours4 trước đó
print(
    "Số equivalence class chưa đạt l=2:",
    len(l2_violating_groups_hours3)
)

print(
    "Số bản ghi nằm trong các nhóm chưa đạt l=2:",
    so_ban_ghi_vi_pham_l2_hours3
)

print(
    "Tỷ lệ bản ghi nằm trong các nhóm chưa đạt l=2:",
    round(ty_le_ban_ghi_vi_pham_l2_hours3, 2),
    "%"
)
```

Như vậy sau khi chuyển từ Hours4 sang Hours3, số bản ghi vi phạm l=2 đã giảm từ 5,671 xuống 4,628, nhưng vấn đề vẫn gần như toàn bộ nằm ở nhóm <=50K.

Phân tích lại QI nào đang tập trung vi phạm l=2 ở strategy Hours3

```python
# Tạo danh sách để lưu kết quả phân tích rủi ro l=2 của từng QI
l2_risk_summary_hours3 = []


# Lần lượt phân tích từng QI trong strategy Country2_Hours3_AgeHigh
for qi_column in candidate_qi_country2_hours3_agehigh:

    # Đếm tổng số bản ghi của từng nhóm giá trị QI
    # trên dataset Hours3 sau khi đã suppression để đạt k=5
    total_by_group = (
        data_country2_hours3_suppressed
        .groupby(qi_column, dropna=False)
        .size()
        .rename("tong_ban_ghi")
    )

    # Đếm số bản ghi chưa đạt l=2 trong từng nhóm giá trị QI
    violation_by_group = (
        violating_l2_records_hours3
        .groupby(qi_column, dropna=False)
        .size()
        .rename("so_ban_ghi_vi_pham_l2")
    )

    # Ghép tổng số bản ghi và số bản ghi vi phạm vào cùng một bảng
    l2_risk_table = (
        pd.concat(
            [total_by_group, violation_by_group],
            axis=1
        )
        .fillna(0)
    )

    # Tính tỷ lệ vi phạm l=2 của từng nhóm giá trị
    l2_risk_table["ty_le_vi_pham_l2_%"] = (
        l2_risk_table["so_ban_ghi_vi_pham_l2"]
        / l2_risk_table["tong_ban_ghi"]
        * 100
    )

    # Tìm nhóm có tỷ lệ vi phạm l=2 cao nhất của QI hiện tại
    highest_risk_group = (
        l2_risk_table
        .sort_values(
            "ty_le_vi_pham_l2_%",
            ascending=False
        )
        .iloc[0]
    )

    # Lưu nhóm rủi ro cao nhất của QI hiện tại
    l2_risk_summary_hours3.append(
        {
            "QI": qi_column,
            "nhom_rui_ro_cao_nhat": highest_risk_group.name,
            "tong_ban_ghi_nhom": int(
                highest_risk_group["tong_ban_ghi"]
            ),
            "so_ban_ghi_vi_pham_l2": int(
                highest_risk_group["so_ban_ghi_vi_pham_l2"]
            ),
            "ty_le_vi_pham_l2_%": round(
                highest_risk_group["ty_le_vi_pham_l2_%"],
                2
            )
        }
    )


# Chuyển kết quả thành DataFrame
l2_risk_summary_hours3 = pd.DataFrame(
    l2_risk_summary_hours3
)


# Sắp xếp từ nhóm có tỷ lệ vi phạm l=2 cao nhất xuống thấp nhất
l2_risk_summary_hours3 = (
    l2_risk_summary_hours3
    .sort_values(
        "ty_le_vi_pham_l2_%",
        ascending=False
    )
    .reset_index(drop=True)
)


# Hiển thị bảng kết quả
l2_risk_summary_hours3
```

Kết quả này cho thấy bước tiếp theo hợp lý nhất không phải gộp Khong_ro của nghề nghiệp/loại công việc, dù tỷ lệ vi phạm cao nhất là 56.18%, vì Khong_ro là dữ liệu thiếu — gộp nó vào một nghề cụ thể sẽ làm sai ý nghĩa dữ liệu.

```python
# Tạo quy tắc generalization học vấn từ 4 nhóm xuống còn 3 nhóm
education_map_3 = {
    "Chua_tot_nghiep_THPT": "THPT_tro_xuong",
    "Tot_nghiep_THPT": "THPT_tro_xuong",
    "Sau_THPT_Cao_dang": "Sau_THPT_Cao_dang",
    "Dai_hoc_tro_len": "Dai_hoc_tro_len"
}


# Tạo cột học vấn mới
# Không ghi đè hoc_van_nhom_4 để còn so sánh strategy cũ và mới
data_anonymized["hoc_van_nhom_3"] = (
    data_anonymized["hoc_van_nhom_4"]
    .map(education_map_3)
)


# Kiểm tra số lượng bản ghi trong từng nhóm học vấn mới
data_anonymized["hoc_van_nhom_3"].value_counts(dropna=False)
```

```python
# Tạo bộ QI cho strategy mới
# Giữ nguyên Country2, Hours3 và AgeHigh
# Chỉ thay hoc_van_nhom_4 bằng hoc_van_nhom_3
candidate_qi_country2_hours3_agehigh_edu3 = [
    "tuoi_nhom_high",
    "loai_cong_viec_nhom",
    "hoc_van_nhom_3",
    "hon_nhan_nhom_2",
    "nghe_nghiep_nhom_4",
    "gio_lam_nhom_3",
    "quoc_gia_nhom_2"
]


# Đánh giá k-anonymity của strategy mới trên toàn bộ dữ liệu
# Chưa suppression để xem riêng generalization học vấn cải thiện được bao nhiêu
privacy_country2_hours3_agehigh_edu3 = evaluate_privacy(
    data_anonymized,
    candidate_qi_country2_hours3_agehigh_edu3,
    k_required=5
)


# Hiển thị kết quả đánh giá privacy của strategy mới
privacy_country2_hours3_agehigh_edu3
```

```python
# Tính kích thước equivalence class tương ứng cho từng bản ghi
# theo strategy Country2_Hours3_AgeHigh_Edu3
group_size_per_record_edu3 = (
    data_anonymized
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )["tuoi"]
    .transform("size")
)


# Đánh dấu các bản ghi đang nằm trong nhóm có kích thước nhỏ hơn k=5
violation_mask_edu3 = (
    group_size_per_record_edu3 < k_required
)


# Tạo dataset mới sau suppression
# Không sửa data_anonymized để còn dùng so sánh các strategy khác
data_country2_hours3_edu3_suppressed = (
    data_anonymized
    .loc[~violation_mask_edu3]
    .copy()
)


# Tính số bản ghi bị suppression
so_suppression_edu3 = (
    len(data_anonymized)
    - len(data_country2_hours3_edu3_suppressed)
)


# Hiển thị kết quả suppression của strategy mới
print("Số bản ghi ban đầu:", len(data_anonymized))
print("Số bản ghi bị suppression:", so_suppression_edu3)
print(
    "Số bản ghi còn lại:",
    len(data_country2_hours3_edu3_suppressed)
)
print(
    "Tỷ lệ suppression:",
    round(
        so_suppression_edu3 / len(data_anonymized) * 100,
        2
    ),
    "%"
)
```

```python
# Đặt ngưỡng l-diversity cần kiểm tra
# Vì thu_nhap có 2 giá trị nên tiếp tục dùng l = 2
l_required = 2


# Với mỗi equivalence class của strategy Edu3,
# đếm số giá trị thu_nhap khác nhau
l_values_edu3 = (
    data_country2_hours3_edu3_suppressed
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )["thu_nhap"]
    .nunique(dropna=False)
)


# Tìm giá trị l nhỏ nhất trong toàn bộ dataset
l_min_edu3 = int(
    l_values_edu3.min()
)


# Đếm số equivalence class chưa đạt l = 2
so_nhom_vi_pham_l2_edu3 = int(
    (l_values_edu3 < l_required).sum()
)


# Đếm số equivalence class có l = 1 hoặc l = 2
l_distribution_edu3 = (
    l_values_edu3
    .value_counts()
    .sort_index()
)


# Hiển thị kết quả kiểm tra l-diversity
print("l_min:", l_min_edu3)

print(
    "Số equivalence class chưa đạt l=2:",
    so_nhom_vi_pham_l2_edu3
)

print("\nPhân bố l-diversity:")
print(l_distribution_edu3)
```

```python
# Tính số bản ghi của từng equivalence class
# trong strategy Edu3 sau khi đã suppression để đạt k=5
group_sizes_edu3_after_suppression = (
    data_country2_hours3_edu3_suppressed
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )
    .size()
)


# Tạo bảng gồm kích thước nhóm và giá trị l-diversity
# để biết mỗi nhóm vi phạm đang chứa bao nhiêu bản ghi
l_group_analysis_edu3 = pd.DataFrame({
    "group_size": group_sizes_edu3_after_suppression,
    "l_value": l_values_edu3
})


# Chỉ lấy các equivalence class chưa đạt l=2
l2_violating_groups_edu3 = (
    l_group_analysis_edu3[
        l_group_analysis_edu3["l_value"] < l_required
    ]
)


# Cộng tổng số bản ghi nằm trong các nhóm chưa đạt l=2
so_ban_ghi_vi_pham_l2_edu3 = int(
    l2_violating_groups_edu3["group_size"].sum()
)


# Tính tỷ lệ bản ghi nằm trong các nhóm chưa đạt l=2
ty_le_ban_ghi_vi_pham_l2_edu3 = (
    so_ban_ghi_vi_pham_l2_edu3
    / len(data_country2_hours3_edu3_suppressed)
    * 100
)


# Hiển thị kết quả
print(
    "Số equivalence class chưa đạt l=2:",
    len(l2_violating_groups_edu3)
)

print(
    "Số bản ghi nằm trong các nhóm chưa đạt l=2:",
    so_ban_ghi_vi_pham_l2_edu3
)

print(
    "Tỷ lệ bản ghi nằm trong các nhóm chưa đạt l=2:",
    round(ty_le_ban_ghi_vi_pham_l2_edu3, 2),
    "%"
)
```

```python
# Tính giá trị l-diversity tương ứng cho từng bản ghi
# Các bản ghi cùng một equivalence class sẽ có cùng giá trị l
l_value_per_record_edu3 = (
    data_country2_hours3_edu3_suppressed
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )["thu_nhap"]
    .transform("nunique")
)


# Đánh dấu những bản ghi nằm trong equivalence class chưa đạt l=2
l2_violation_mask_edu3 = (
    l_value_per_record_edu3 < l_required
)


# Lấy riêng các bản ghi chưa đạt l=2
# Dùng copy để không làm thay đổi dataset đã đạt k=5
violating_l2_records_edu3 = (
    data_country2_hours3_edu3_suppressed
    .loc[l2_violation_mask_edu3]
    .copy()
)


# Gắn giá trị l vào từng bản ghi để tiện phân tích ở bước sau
violating_l2_records_edu3["l_value"] = (
    l_value_per_record_edu3[
        l2_violation_mask_edu3
    ]
)


# Kiểm tra số lượng bản ghi chưa đạt l=2
print(
    "Số bản ghi chưa đạt l=2 của Edu3:",
    len(violating_l2_records_edu3)
)


# Kiểm tra các bản ghi vi phạm đang thuộc loại thu nhập nào
print("\nPhân bố thu nhập:")
print(
    violating_l2_records_edu3["thu_nhap"]
    .value_counts()
)
```

```python
# Tạo danh sách để lưu kết quả phân tích rủi ro l=2 của từng QI
l2_risk_summary_edu3 = []


# Lần lượt phân tích từng QI trong strategy Edu3 hiện tại
for qi_column in candidate_qi_country2_hours3_agehigh_edu3:

    # Đếm tổng số bản ghi của từng nhóm giá trị QI
    # trên dataset Edu3 sau khi suppression để đạt k=5
    total_by_group = (
        data_country2_hours3_edu3_suppressed
        .groupby(qi_column, dropna=False)
        .size()
        .rename("tong_ban_ghi")
    )

    # Đếm số bản ghi chưa đạt l=2 trong từng nhóm giá trị QI
    violation_by_group = (
        violating_l2_records_edu3
        .groupby(qi_column, dropna=False)
        .size()
        .rename("so_ban_ghi_vi_pham_l2")
    )

    # Ghép tổng số bản ghi và số bản ghi vi phạm vào cùng một bảng
    l2_risk_table = (
        pd.concat(
            [total_by_group, violation_by_group],
            axis=1
        )
        .fillna(0)
    )

    # Tính tỷ lệ bản ghi vi phạm l=2 trong từng nhóm giá trị
    l2_risk_table["ty_le_vi_pham_l2_%"] = (
        l2_risk_table["so_ban_ghi_vi_pham_l2"]
        / l2_risk_table["tong_ban_ghi"]
        * 100
    )

    # Tìm nhóm giá trị có tỷ lệ vi phạm l=2 cao nhất của QI hiện tại
    highest_risk_group = (
        l2_risk_table
        .sort_values(
            "ty_le_vi_pham_l2_%",
            ascending=False
        )
        .iloc[0]
    )

    # Lưu nhóm rủi ro cao nhất của QI hiện tại vào danh sách
    l2_risk_summary_edu3.append(
        {
            "QI": qi_column,
            "nhom_rui_ro_cao_nhat": highest_risk_group.name,
            "tong_ban_ghi_nhom": int(
                highest_risk_group["tong_ban_ghi"]
            ),
            "so_ban_ghi_vi_pham_l2": int(
                highest_risk_group["so_ban_ghi_vi_pham_l2"]
            ),
            "ty_le_vi_pham_l2_%": round(
                highest_risk_group["ty_le_vi_pham_l2_%"],
                2
            )
        }
    )


# Chuyển kết quả thành DataFrame để dễ xem
l2_risk_summary_edu3 = pd.DataFrame(
    l2_risk_summary_edu3
)


# Sắp xếp từ nhóm có tỷ lệ vi phạm l=2 cao nhất xuống thấp nhất
l2_risk_summary_edu3 = (
    l2_risk_summary_edu3
    .sort_values(
        "ty_le_vi_pham_l2_%",
        ascending=False
    )
    .reset_index(drop=True)
)


# Hiển thị bảng tổng hợp rủi ro l=2 của strategy Edu3
l2_risk_summary_edu3
```

```python
# Đánh giá lại k-anonymity trên dataset sau khi suppression cả k=5 và l=2
privacy_final_candidate = evaluate_privacy(
    data_country2_hours3_age3_edu3_l2_suppressed,
    candidate_qi_country2_hours3_age3_edu3,
    k_required=5
)


# Với mỗi equivalence class còn lại,
# đếm số giá trị thu_nhap khác nhau để kiểm tra l-diversity
l_values_final_candidate = (
    data_country2_hours3_age3_edu3_l2_suppressed
    .groupby(
        candidate_qi_country2_hours3_age3_edu3,
        dropna=False
    )["thu_nhap"]
    .nunique(dropna=False)
)


# Tìm giá trị l nhỏ nhất của dataset sau cùng
l_min_final_candidate = int(
    l_values_final_candidate.min()
)


# Đếm số equivalence class còn vi phạm l=2
so_nhom_vi_pham_l2_final = int(
    (l_values_final_candidate < 2).sum()
)


# Hiển thị kết quả k-anonymity
print("KẾT QUẢ K-ANONYMITY:")
print("k_min:", privacy_final_candidate["k_min"])
print(
    "Số bản ghi vi phạm k=5:",
    privacy_final_candidate["so_ban_ghi_vi_pham_k5"]
)


# Hiển thị kết quả l-diversity
print("\nKẾT QUẢ L-DIVERSITY:")
print("l_min:", l_min_final_candidate)
print(
    "Số equivalence class vi phạm l=2:",
    so_nhom_vi_pham_l2_final
)
```

```python
# Tính lại giá trị l-diversity cho từng bản ghi
# của strategy Country2 + Hours3 + AgeHigh + Edu3 sau khi đã đạt k=5
l_value_per_record_strategy2 = (
    data_country2_hours3_edu3_suppressed
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )["thu_nhap"]
    .transform("nunique")
)


# Đánh dấu các bản ghi nằm trong equivalence class chưa đạt l=2
l2_violation_mask_strategy2 = (
    l_value_per_record_strategy2 < 2
)


# Tạo dữ liệu cuối của Strategy 2
# Loại nguyên các equivalence class còn vi phạm l=2
data_strategy2_k5_l2 = (
    data_country2_hours3_edu3_suppressed
    .loc[~l2_violation_mask_strategy2]
    .copy()
)


# Tính số bản ghi phải suppression thêm để đạt l=2
so_suppression_them_strategy2 = int(
    l2_violation_mask_strategy2.sum()
)


# Tính tổng số bản ghi bị suppression so với dữ liệu ban đầu
tong_suppression_strategy2 = (
    len(data_anonymized)
    - len(data_strategy2_k5_l2)
)


# Tính tỷ lệ suppression tổng của Strategy 2
ty_le_suppression_strategy2 = (
    tong_suppression_strategy2
    / len(data_anonymized)
    * 100
)


# Hiển thị kết quả để kiểm tra Strategy 2
print(
    "Suppression thêm để đạt l=2:",
    so_suppression_them_strategy2
)

print(
    "Tổng số bản ghi suppression:",
    tong_suppression_strategy2
)

print(
    "Số bản ghi còn lại:",
    len(data_strategy2_k5_l2)
)

print(
    "Tỷ lệ suppression tổng:",
    round(ty_le_suppression_strategy2, 2),
    "%"
)
```

```python
# Đánh giá lại k-anonymity trên dữ liệu cuối của Strategy 2
privacy_strategy2 = evaluate_privacy(
    data_strategy2_k5_l2,
    candidate_qi_country2_hours3_agehigh_edu3,
    k_required=5
)


# Với mỗi equivalence class của Strategy 2,
# đếm số giá trị thu_nhap khác nhau để kiểm tra l-diversity
l_values_strategy2 = (
    data_strategy2_k5_l2
    .groupby(
        candidate_qi_country2_hours3_agehigh_edu3,
        dropna=False
    )["thu_nhap"]
    .nunique(dropna=False)
)


# Tìm giá trị l nhỏ nhất của toàn bộ Strategy 2
l_min_strategy2 = int(
    l_values_strategy2.min()
)


# Đếm số equivalence class còn vi phạm l=2
so_nhom_vi_pham_l2_strategy2 = int(
    (l_values_strategy2 < 2).sum()
)


# Hiển thị kết quả kiểm tra k-anonymity
print("KẾT QUẢ K-ANONYMITY - STRATEGY 2")
print(
    "k_min:",
    privacy_strategy2["k_min"]
)
print(
    "Số bản ghi vi phạm k=5:",
    privacy_strategy2["so_ban_ghi_vi_pham_k5"]
)


# Hiển thị kết quả kiểm tra l-diversity
print("\nKẾT QUẢ L-DIVERSITY - STRATEGY 2")
print(
    "l_min:",
    l_min_strategy2
)
print(
    "Số equivalence class vi phạm l=2:",
    so_nhom_vi_pham_l2_strategy2
)
```

Strategy 2 đạt cả k-anonymity với k=5 và l-diversity với l=2. Không còn bản ghi vi phạm k=5 và không còn equivalence class vi phạm l=2. Do đó Strategy 2 đáp ứng các điều kiện Privacy đã đặt ra.

```python
# Đếm số lượng của từng giá trị trong cột nghe_nghiep
# Mục đích là xem chính xác dataset đang dùng những tên nghề nghiệp nào
occupation_counts = (
    data_check["nghe_nghiep"]
    .value_counts(
        dropna=False
    )
)


# Hiển thị toàn bộ nghề nghiệp và số bản ghi tương ứng
# Ta sẽ dựa đúng vào các tên này để viết hàm generalize_occupation_4()
print(
    occupation_counts
)
```

```python
# Import importlib để có thể nạp lại phiên bản mới nhất của anonymization_engine.py
import importlib


# Import module anonymization_engine chứa các hàm ẩn danh hóa
import anonymization_engine


# Nạp lại module để notebook nhận hàm apply_strategy_1_generalization() vừa thêm
importlib.reload(
    anonymization_engine
)


# Áp dụng toàn bộ 7 phép generalization của Strategy 1
# lên dataset gốc đang lưu trong biến data_check
generalized_strategy1_check = (
    anonymization_engine.apply_strategy_1_generalization(
        data_check
    )
)


# Hiển thị kích thước dataset gốc
# Dataset gốc của chúng ta có 15 cột
print(
    "Kích thước dữ liệu gốc:",
    data_check.shape
)


# Hiển thị kích thước sau khi generalization
# Vì hàm tạo thêm 7 cột mới nên dự kiến sẽ có 22 cột
print(
    "Kích thước sau generalization:",
    generalized_strategy1_check.shape
)


# Chọn các cột gốc và cột sau generalization để đối chiếu trực tiếp
columns_to_check = [
    "tuoi",
    "tuoi_nhom_3",
    "loai_cong_viec",
    "loai_cong_viec_nhom",
    "so_nam_hoc",
    "hoc_van_nhom_3",
    "tinh_trang_hon_nhan",
    "hon_nhan_nhom_2",
    "nghe_nghiep",
    "nghe_nghiep_nhom_4",
    "gio_lam_moi_tuan",
    "gio_lam_nhom_3",
    "quoc_gia",
    "quoc_gia_nhom_2"
]


# Hiển thị 10 bản ghi đầu tiên
# để kiểm tra giá trị gốc đã được chuyển sang nhóm mới như thế nào
display(
    generalized_strategy1_check[
        columns_to_check
    ].head(
        10
    )
)
```

```python
# Import importlib để nạp lại phiên bản mới nhất của anonymization_engine.py
import importlib


# Import module chứa các hàm ẩn danh hóa
import anonymization_engine


# Nạp lại module vì chúng ta vừa sửa hàm generalize_country_2()
importlib.reload(
    anonymization_engine
)


# Áp dụng 7 phép generalization của Strategy 1 lên dataset gốc
# Kết quả được lưu vào biến riêng để kiểm tra, chưa ghi ra file
generalized_strategy1_check = (
    anonymization_engine.apply_strategy_1_generalization(
        data_check
    )
)


# Đếm số bản ghi của từng nhóm quốc gia sau generalization
# Mục đích là kiểm tra Lanh_tho_hai_ngoai_My đã được nhập đúng vào Hoa_Ky
country2_counts = (
    generalized_strategy1_check["quoc_gia_nhom_2"]
    .value_counts(
        dropna=False
    )
)


# Hiển thị kết quả kiểm tra
print(
    country2_counts
)
```

```python
# Import importlib để có thể nạp lại module privacy_analyzer mới nhất
import importlib


# Import module privacy_analyzer chứa hàm đánh giá k-anonymity
import privacy_analyzer


# Nạp lại module để notebook sử dụng phiên bản code hiện tại
importlib.reload(
    privacy_analyzer
)


# Khai báo 7 QI sau generalization của Strategy 1
# Đây chính là các thuộc tính dùng để tạo equivalence class
strategy1_qi_columns = [
    "tuoi_nhom_3",
    "loai_cong_viec_nhom",
    "hoc_van_nhom_3",
    "hon_nhan_nhom_2",
    "nghe_nghiep_nhom_4",
    "gio_lam_nhom_3",
    "quoc_gia_nhom_2"
]


# Đánh giá mức độ k-anonymity của dữ liệu sau generalization
# k_required=5 nghĩa là mỗi equivalence class cần có ít nhất 5 bản ghi
strategy1_privacy_before_suppression = (
    privacy_analyzer.evaluate_privacy(
        generalized_strategy1_check,
        strategy1_qi_columns,
        k_required=5
    )
)


# Hiển thị kết quả đánh giá privacy trước khi suppression
print(
    strategy1_privacy_before_suppression
)
```

```python
# Import importlib để có thể nạp lại phiên bản mới nhất của anonymization_engine.py
import importlib


# Import module anonymization_engine chứa hàm suppression vừa thêm
import anonymization_engine


# Nạp lại module để notebook nhận hàm suppress_k_anonymity() mới nhất
importlib.reload(
    anonymization_engine
)


# Thực hiện suppression các bản ghi nằm trong equivalence class có kích thước < 5
# Dữ liệu đầu vào là Strategy 1 sau generalization
strategy1_after_k_suppression = (
    anonymization_engine.suppress_k_anonymity(
        generalized_strategy1_check,
        strategy1_qi_columns,
        k_required=5
    )
)


# Lấy số bản ghi trước khi suppression
so_ban_ghi_truoc_k = len(
    generalized_strategy1_check
)


# Lấy số bản ghi còn lại sau khi suppression
so_ban_ghi_sau_k = len(
    strategy1_after_k_suppression
)


# Tính số bản ghi đã bị suppression
so_ban_ghi_suppression_k = (
    so_ban_ghi_truoc_k
    - so_ban_ghi_sau_k
)


# Hiển thị số bản ghi trước suppression
print(
    "Số bản ghi trước suppression:",
    so_ban_ghi_truoc_k
)


# Hiển thị số bản ghi bị suppression để đạt k=5
print(
    "Số bản ghi suppression k=5:",
    so_ban_ghi_suppression_k
)


# Hiển thị số bản ghi còn lại
print(
    "Số bản ghi còn lại:",
    so_ban_ghi_sau_k
)
```

```python
# Đánh giá lại k-anonymity sau khi đã suppression
# Mục đích là xác nhận không còn equivalence class nào có kích thước nhỏ hơn 5
strategy1_privacy_after_k = (
    privacy_analyzer.evaluate_privacy(
        strategy1_after_k_suppression,
        strategy1_qi_columns,
        k_required=5
    )
)


# Hiển thị toàn bộ kết quả đánh giá sau suppression
print(
    strategy1_privacy_after_k
)
```

```python
# Đánh giá l-diversity trên dữ liệu đã đạt k=5
# Mục đích là kiểm tra trong mỗi equivalence class
# có đủ ít nhất 2 giá trị thu_nhap khác nhau hay không
strategy1_l_diversity_after_k = (
    privacy_analyzer.evaluate_l_diversity(
        strategy1_after_k_suppression,
        strategy1_qi_columns,
        sensitive_column="thu_nhap",
        l_required=2
    )
)


# Hiển thị toàn bộ kết quả đánh giá l-diversity
print(
    strategy1_l_diversity_after_k
)
```

```python
# Import importlib để có thể nạp lại phiên bản mới nhất của anonymization_engine.py
import importlib


# Import module anonymization_engine chứa hàm suppress_l_diversity() vừa thêm
import anonymization_engine


# Nạp lại module để notebook nhận code mới nhất
importlib.reload(
    anonymization_engine
)


# Thực hiện suppression các equivalence class chưa đạt l=2
# Dữ liệu đầu vào là dữ liệu đã đạt k=5
strategy1_after_l_suppression = (
    anonymization_engine.suppress_l_diversity(
        strategy1_after_k_suppression,
        strategy1_qi_columns,
        sensitive_column="thu_nhap",
        l_required=2
    )
)


# Lấy số bản ghi trước khi l-suppression
so_ban_ghi_truoc_l = len(
    strategy1_after_k_suppression
)


# Lấy số bản ghi còn lại sau l-suppression
so_ban_ghi_sau_l = len(
    strategy1_after_l_suppression
)


# Tính số bản ghi bị suppression thêm vì chưa đạt l=2
so_ban_ghi_suppression_l = (
    so_ban_ghi_truoc_l
    - so_ban_ghi_sau_l
)


# Hiển thị số bản ghi trước l-suppression
print(
    "Số bản ghi trước l-suppression:",
    so_ban_ghi_truoc_l
)


# Hiển thị số bản ghi bị suppression thêm vì l=2
print(
    "Số bản ghi suppression thêm vì l=2:",
    so_ban_ghi_suppression_l
)


# Hiển thị số bản ghi còn lại cuối cùng
print(
    "Số bản ghi còn lại:",
    so_ban_ghi_sau_l
)
```

```python
# Đánh giá đồng thời k-anonymity và l-diversity
# trên dữ liệu sau khi đã thực hiện cả k-suppression và l-suppression
strategy1_final_privacy = (
    privacy_analyzer.evaluate_privacy_full(

        # Dữ liệu cuối cùng hiện có 30.604 bản ghi
        strategy1_after_l_suppression,

        # 7 cột QI của Strategy 1
        strategy1_qi_columns,

        # Cột thu_nhap là thuộc tính nhạy cảm dùng để kiểm tra l-diversity
        "thu_nhap",

        # Yêu cầu mỗi equivalence class phải có ít nhất 5 bản ghi
        k_required=5,

        # Yêu cầu mỗi equivalence class phải có ít nhất 2 giá trị thu_nhap khác nhau
        l_required=2
    )
)


# Hiển thị toàn bộ kết quả kiểm tra Privacy cuối cùng
print(
    strategy1_final_privacy
)
```

```python
# Tạo danh sách các cột được phép giữ lại
# trong dữ liệu cuối cùng sau khi ẩn danh hóa
release_columns_strategy1 = [

    # Nhóm tuổi đã được generalization
    "tuoi_nhom_3",

    # Loại công việc đã được generalization
    "loai_cong_viec_nhom",

    # Học vấn đã được generalization
    "hoc_van_nhom_3",

    # Tình trạng hôn nhân đã được generalization
    "hon_nhan_nhom_2",

    # Nghề nghiệp đã được generalization
    "nghe_nghiep_nhom_4",

    # Giờ làm việc đã được generalization
    "gio_lam_nhom_3",

    # Quốc gia đã được generalization
    "quoc_gia_nhom_2",

    # Giữ lại thu_nhap vì đây là thuộc tính nhạy cảm
    # đồng thời là target dùng cho Machine Learning
    "thu_nhap"
]


# Chỉ lấy 8 cột được phép phát hành
# từ dữ liệu đã đạt cả k=5 và l=2
strategy1_release = (
    strategy1_after_l_suppression[
        release_columns_strategy1
    ]
    .copy()
)


# Hiển thị kích thước của dữ liệu phát hành
# Kỳ vọng kết quả là 30.604 dòng và 8 cột
print(
    "Kích thước dữ liệu phát hành:",
    strategy1_release.shape
)


# Hiển thị chính xác tên các cột còn lại
# để chắc chắn không còn cột dữ liệu gốc chi tiết
print(
    "Các cột dữ liệu phát hành:",
    strategy1_release.columns.tolist()
)


# Hiển thị 5 bản ghi đầu tiên
# để kiểm tra trực quan dữ liệu cuối cùng
display(
    strategy1_release.head()
)
```

```python
# Import pandas để đọc lại file Strategy 1 đã lưu trước đây
import pandas as pd


# Đọc file dữ liệu ẩn danh Strategy 1 cũ từ ổ đĩa
# Mục đích là dùng file này làm kết quả tham chiếu để đối chiếu
strategy1_saved_old = pd.read_csv(
    "../data/processed_da_xu_li/anonymized_candidate_k5_l2.csv",
    encoding="utf-8-sig"
)


# Reset index của dữ liệu vừa tạo bằng anonymization_engine
# Vì dữ liệu hiện tại vẫn giữ index gốc để phục vụ Machine Learning
strategy1_release_for_compare = (
    strategy1_release
    .reset_index(
        drop=True
    )
)


# Reset index của file cũ để hai DataFrame có cùng kiểu index khi so sánh
strategy1_saved_old_for_compare = (
    strategy1_saved_old
    .reset_index(
        drop=True
    )
)


# So sánh toàn bộ DataFrame:
# - số dòng
# - số cột
# - tên cột
# - thứ tự dòng
# - giá trị của từng ô
# Nếu tất cả giống nhau thì kết quả sẽ là True
same_as_saved_strategy1 = (
    strategy1_release_for_compare
    .equals(
        strategy1_saved_old_for_compare
    )
)


# Hiển thị kích thước của dữ liệu vừa tái tạo
print(
    "Kích thước dữ liệu mới:",
    strategy1_release_for_compare.shape
)


# Hiển thị kích thước của file Strategy 1 cũ
print(
    "Kích thước dữ liệu cũ:",
    strategy1_saved_old_for_compare.shape
)


# Hiển thị kết quả so sánh toàn bộ hai DataFrame
print(
    "Hai bộ dữ liệu có giống hoàn toàn không?:",
    same_as_saved_strategy1
)
```

```python
# Import importlib để nạp lại phiên bản mới nhất của anonymization_engine.py
import importlib


# Import module chứa các hàm ẩn danh hóa
import anonymization_engine


# Nạp lại module vì chúng ta vừa thêm apply_strategy_2_generalization()
importlib.reload(
    anonymization_engine
)


# Áp dụng toàn bộ generalization của Strategy 2 lên dataset gốc
# Strategy 2 dùng tuoi_nhom_high thay cho tuoi_nhom_3
strategy2_generalized_check = (
    anonymization_engine.apply_strategy_2_generalization(
        data_check
    )
)


# Khai báo 7 cột QI của Strategy 2
# Điểm khác Strategy 1 nằm ở cột tuoi_nhom_high
strategy2_qi_columns = [
    "tuoi_nhom_high",
    "loai_cong_viec_nhom",
    "hoc_van_nhom_3",
    "hon_nhan_nhom_2",
    "nghe_nghiep_nhom_4",
    "gio_lam_nhom_3",
    "quoc_gia_nhom_2"
]


# Đánh giá k-anonymity của Strategy 2 trước khi suppression
# Mục đích là kiểm tra code mới có tái tạo đúng Strategy 2 cũ hay không
strategy2_privacy_before_suppression = (
    privacy_analyzer.evaluate_privacy(
        strategy2_generalized_check,
        strategy2_qi_columns,
        k_required=5
    )
)


# Hiển thị kích thước dữ liệu sau generalization
# Vì thêm 7 cột mới nên dự kiến là 32.561 dòng và 22 cột
print(
    "Kích thước sau generalization Strategy 2:",
    strategy2_generalized_check.shape
)


# Hiển thị toàn bộ kết quả k-anonymity trước suppression
print(
    strategy2_privacy_before_suppression
)
```

```python
# Thực hiện suppression các bản ghi của Strategy 2
# nằm trong equivalence class có kích thước nhỏ hơn 5
strategy2_after_k_suppression = (
    anonymization_engine.suppress_k_anonymity(

        # Dữ liệu Strategy 2 sau generalization
        strategy2_generalized_check,

        # 7 cột QI của Strategy 2
        strategy2_qi_columns,

        # Yêu cầu mỗi equivalence class phải có ít nhất 5 bản ghi
        k_required=5
    )
)


# Lấy tổng số bản ghi trước khi thực hiện k-suppression
so_ban_ghi_truoc_k_strategy2 = len(
    strategy2_generalized_check
)


# Lấy tổng số bản ghi còn lại sau khi thực hiện k-suppression
so_ban_ghi_sau_k_strategy2 = len(
    strategy2_after_k_suppression
)


# Tính số bản ghi đã bị suppression vì chưa đạt k=5
so_ban_ghi_suppression_k_strategy2 = (
    so_ban_ghi_truoc_k_strategy2
    - so_ban_ghi_sau_k_strategy2
)


# Hiển thị số bản ghi trước suppression
print(
    "Số bản ghi trước suppression:",
    so_ban_ghi_truoc_k_strategy2
)


# Hiển thị số bản ghi bị suppression vì chưa đạt k=5
print(
    "Số bản ghi suppression k=5:",
    so_ban_ghi_suppression_k_strategy2
)


# Hiển thị số bản ghi còn lại sau k-suppression
print(
    "Số bản ghi còn lại:",
    so_ban_ghi_sau_k_strategy2
)
```

```python
# Đánh giá lại k-anonymity của Strategy 2
# sau khi đã loại các bản ghi thuộc nhóm có kích thước nhỏ hơn 5
strategy2_privacy_after_k = (
    privacy_analyzer.evaluate_privacy(

        # Dữ liệu Strategy 2 sau k-suppression
        strategy2_after_k_suppression,

        # 7 cột QI của Strategy 2
        strategy2_qi_columns,

        # Yêu cầu k tối thiểu bằng 5
        k_required=5
    )
)


# Hiển thị toàn bộ kết quả kiểm tra sau k-suppression
print(
    strategy2_privacy_after_k
)
```

```python
# Đánh giá l-diversity trên dữ liệu Strategy 2
# sau khi dữ liệu đã đạt k-anonymity với k=5
strategy2_l_diversity_after_k = (
    privacy_analyzer.evaluate_l_diversity(

        # Dữ liệu Strategy 2 sau k-suppression
        strategy2_after_k_suppression,

        # 7 cột QI của Strategy 2
        strategy2_qi_columns,

        # Cột thu_nhap là thuộc tính nhạy cảm
        # cần được bảo vệ khỏi việc suy luận trong từng equivalence class
        sensitive_column="thu_nhap",

        # Yêu cầu mỗi equivalence class
        # phải có ít nhất 2 giá trị thu_nhap khác nhau
        l_required=2
    )
)


# Hiển thị toàn bộ kết quả đánh giá l-diversity
print(
    strategy2_l_diversity_after_k
)
```

```python
# Thực hiện suppression các equivalence class của Strategy 2
# chưa đạt yêu cầu l-diversity với l=2
strategy2_after_l_suppression = (
    anonymization_engine.suppress_l_diversity(

        # Dữ liệu Strategy 2 đã đạt k=5
        strategy2_after_k_suppression,

        # 7 cột QI của Strategy 2
        strategy2_qi_columns,

        # Cột thu_nhap là thuộc tính nhạy cảm
        # dùng để kiểm tra l-diversity
        sensitive_column="thu_nhap",

        # Yêu cầu mỗi equivalence class
        # phải có ít nhất 2 giá trị thu_nhap khác nhau
        l_required=2
    )
)


# Lấy số bản ghi trước khi thực hiện l-suppression
so_ban_ghi_truoc_l_strategy2 = len(
    strategy2_after_k_suppression
)


# Lấy số bản ghi còn lại sau khi thực hiện l-suppression
so_ban_ghi_sau_l_strategy2 = len(
    strategy2_after_l_suppression
)


# Tính số bản ghi bị suppression thêm
# do nằm trong equivalence class chưa đạt l=2
so_ban_ghi_suppression_l_strategy2 = (
    so_ban_ghi_truoc_l_strategy2
    - so_ban_ghi_sau_l_strategy2
)


# Hiển thị số bản ghi trước l-suppression
print(
    "Số bản ghi trước l-suppression:",
    so_ban_ghi_truoc_l_strategy2
)


# Hiển thị số bản ghi bị suppression thêm vì chưa đạt l=2
print(
    "Số bản ghi suppression thêm vì l=2:",
    so_ban_ghi_suppression_l_strategy2
)


# Hiển thị số bản ghi còn lại sau l-suppression
print(
    "Số bản ghi còn lại:",
    so_ban_ghi_sau_l_strategy2
)
```

```python

# TỰ ĐỘNG CHẠY NHIỀU STRATEGY



# Tạo danh sách tên các Strategy mà hệ thống cần chạy
# Sau này nếu có Strategy 3, Strategy 4 thì chỉ cần thêm tên vào đây
strategy_names = [
    "strategy_1",
    "strategy_2"
]


# Tạo dictionary rỗng để lưu kết quả của từng Strategy
# Mỗi Strategy sẽ lưu:
# - dữ liệu ẩn danh
# - index gốc được giữ lại
strategy_results = {}


# Lần lượt lấy từng tên Strategy trong danh sách
for strategy_name in strategy_names:

    # Gọi hàm chung anonymize_by_strategy()
    # để hệ thống tự chạy đúng pipeline tương ứng
    release_data, retained_indices = (
        anonymization_engine.anonymize_by_strategy(

            # Dataset gốc
            data_check,

            # Strategy đang được chạy trong vòng lặp
            strategy_name=strategy_name,

            # Yêu cầu k-anonymity bằng 5
            k_required=5,

            # Yêu cầu l-diversity bằng 2
            l_required=2
        )
    )


    # Lưu kết quả của Strategy hiện tại vào dictionary
    strategy_results[strategy_name] = {

        # Dataset sau ẩn danh hóa
        "release_data": release_data,

        # Index gốc của các bản ghi được giữ lại
        "retained_indices": retained_indices
    }


    # Hiển thị tên Strategy vừa chạy
    print(
        "Strategy:",
        strategy_name
    )


    # Hiển thị kích thước dữ liệu ẩn danh của Strategy đó
    print(
        "Kích thước:",
        release_data.shape
    )


    # Hiển thị số bản ghi còn được giữ lại
    print(
        "Số bản ghi giữ lại:",
        len(
            retained_indices
        )
    )


    # In dòng phân cách để kết quả dễ đọc
    print(
        "----------------------------------------"
    )
```

```python

# TỰ ĐỘNG ĐÁNH GIÁ PRIVACY CHO NHIỀU STRATEGY


# Tạo dictionary để khai báo các cột QI tương ứng
# với từng Strategy
#
# Strategy 1 sử dụng tuoi_nhom_3
# Strategy 2 sử dụng tuoi_nhom_high
strategy_qi_map = {

    # 7 QI của Strategy 1
    "strategy_1": [
        "tuoi_nhom_3",
        "loai_cong_viec_nhom",
        "hoc_van_nhom_3",
        "hon_nhan_nhom_2",
        "nghe_nghiep_nhom_4",
        "gio_lam_nhom_3",
        "quoc_gia_nhom_2"
    ],

    # 7 QI của Strategy 2
    "strategy_2": [
        "tuoi_nhom_high",
        "loai_cong_viec_nhom",
        "hoc_van_nhom_3",
        "hon_nhan_nhom_2",
        "nghe_nghiep_nhom_4",
        "gio_lam_nhom_3",
        "quoc_gia_nhom_2"
    ]
}


# Lần lượt lấy từng Strategy đã chạy trước đó
for strategy_name in strategy_names:

    # Lấy dataset ẩn danh cuối cùng của Strategy hiện tại
    release_data = (
        strategy_results[strategy_name]["release_data"]
    )


    # Lấy đúng danh sách QI tương ứng với Strategy hiện tại
    qi_columns = (
        strategy_qi_map[strategy_name]
    )


    # Đánh giá đồng thời:
    # - k-anonymity
    # - l-diversity
    # - privacy_valid
    privacy_result = (
        privacy_analyzer.evaluate_privacy_full(

            # Dataset ẩn danh cuối cùng
            release_data,

            # Các cột QI của Strategy hiện tại
            qi_columns,

            # Cột thu_nhap là thuộc tính nhạy cảm
            "thu_nhap",

            # Yêu cầu k=5
            k_required=5,

            # Yêu cầu l=2
            l_required=2
        )
    )


    # Lưu kết quả Privacy vào strategy_results
    # để các bước sau có thể dùng tiếp mà không phải tính lại
    strategy_results[strategy_name]["privacy"] = (
        privacy_result
    )


    # Hiển thị tên Strategy đang được đánh giá
    print(
        "Strategy:",
        strategy_name
    )


    # Hiển thị toàn bộ kết quả Privacy
    print(
        privacy_result
    )


    # In dòng phân cách để dễ đọc
    print(
        "----------------------------------------"
    )
```

```python
# ============================================================
# TỰ ĐỘNG TÍNH SUPPRESSION VÀ DATA RETENTION
# CHO TỪNG STRATEGY
# ============================================================


# Lấy tổng số bản ghi của dataset gốc
# Dataset ban đầu có 32.561 bản ghi
total_original_records = len(
    data_check
)


# Lần lượt lấy từng Strategy đã chạy trước đó
for strategy_name in strategy_names:

    # Lấy danh sách index của các bản ghi
    # còn được giữ lại sau quá trình ẩn danh hóa
    retained_indices = (
        strategy_results[strategy_name]["retained_indices"]
    )


    # Đếm số bản ghi còn được giữ lại
    retained_records = len(
        retained_indices
    )


    # Tính số bản ghi đã bị suppression
    # bằng tổng dữ liệu gốc trừ số bản ghi còn lại
    suppressed_records = (
        total_original_records
        - retained_records
    )


    # Tính tỷ lệ suppression theo phần trăm
    # Ví dụ 1.957 / 32.561 * 100 ≈ 6.01%
    suppression_percent = (
        suppressed_records
        / total_original_records
        * 100
    )


    # Tính tỷ lệ dữ liệu còn được giữ lại
    # Data Retention = 100% - Suppression %
    data_retention_percent = (
        retained_records
        / total_original_records
        * 100
    )


    # Lưu số bản ghi bị suppression
    # vào kết quả của Strategy hiện tại
    strategy_results[strategy_name]["suppressed_records"] = (
        suppressed_records
    )


    # Lưu tỷ lệ suppression
    # làm tròn đến 2 chữ số thập phân
    strategy_results[strategy_name]["suppression_percent"] = round(
        suppression_percent,
        2
    )


    # Lưu tỷ lệ dữ liệu được giữ lại
    # làm tròn đến 2 chữ số thập phân
    strategy_results[strategy_name]["data_retention_percent"] = round(
        data_retention_percent,
        2
    )


    # Hiển thị tên Strategy
    print(
        "Strategy:",
        strategy_name
    )


    # Hiển thị số bản ghi bị suppression
    print(
        "Số bản ghi suppression:",
        suppressed_records
    )


    # Hiển thị tỷ lệ suppression
    print(
        "Suppression (%):",
        round(
            suppression_percent,
            2
        )
    )


    # Hiển thị tỷ lệ dữ liệu còn được giữ lại
    print(
        "Data Retention (%):",
        round(
            data_retention_percent,
            2
        )
    )


    # In dòng phân cách để kết quả dễ đọc
    print(
        "----------------------------------------"
    )
```

```python
# ============================================================
# GẮN VALIDATION UTILITY SCORE TỰ ĐỘNG CHO TỪNG STRATEGY
# ============================================================


# Tạo dictionary chứa Utility Score của từng Strategy
# Các giá trị này KHÔNG còn nhập thủ công 97.41 và 97.46
# mà được lấy trực tiếp từ kết quả Machine Learning tự động
validation_utility_scores = {

    # Lấy Utility Score tự động của Strategy 1
    # Giá trị này được Hàm 9 trong ml_evaluator.py tính ra
    "strategy_1": (
        strategy1_ml_auto_result[
            "utility_score_percent"
        ]
    ),

    # Lấy Utility Score tự động của Strategy 2
    # Giá trị này cũng được Hàm 9 tự động tính ra
    "strategy_2": (
        strategy2_ml_auto_result[
            "utility_score_percent"
        ]
    )
}


# Lần lượt duyệt qua từng Strategy
# strategy_names hiện chứa strategy_1 và strategy_2
for strategy_name in strategy_names:

    # Lấy Utility Score tự động
    # tương ứng với Strategy đang được xét
    utility_score = (
        validation_utility_scores[
            strategy_name
        ]
    )


    # Gắn Utility Score vừa tính tự động
    # vào dictionary kết quả của Strategy tương ứng
    #
    # strategy_selector.py sẽ dùng giá trị này
    # để tính Selection Score ở bước sau
    strategy_results[
        strategy_name
    ][
        "utility_score_percent"
    ] = utility_score


    # Hiển thị tên Strategy đang được kiểm tra
    print(
        "Strategy:",
        strategy_name
    )


    # Hiển thị Validation Utility Score
    # làm tròn đến 2 chữ số thập phân
    print(
        "Validation Utility Score (%):",
        round(
            strategy_results[
                strategy_name
            ][
                "utility_score_percent"
            ],
            2
        )
    )


    # In dòng phân cách
    # để kết quả của hai Strategy dễ đọc hơn
    print(
        "----------------------------------------"
    )
```

```python
# ============================================================
# CHUẨN BỊ DỮ LIỆU ĐẦU VÀO CHO STRATEGY SELECTOR
# ============================================================


# Tạo danh sách chứa thông tin cần thiết
# để strategy_selector.py so sánh các Strategy
strategies_for_selection = []


# Duyệt lần lượt từng Strategy
for strategy_name in strategy_names:

    # Lấy toàn bộ kết quả đã lưu
    # của Strategy hiện tại
    current_result = (
        strategy_results[
            strategy_name
        ]
    )


    # Tạo dictionary chứa những thông tin
    # mà Strategy Selector cần sử dụng
    strategy_selection_data = {

        # Tên Strategy
        "strategy": strategy_name,

        # Tỷ lệ suppression của Strategy
        # Giá trị này được tính tự động từ quá trình ẩn danh hóa
        "suppression_percent": (
            current_result[
                "suppression_percent"
            ]
        ),

        # Utility Score Validation
        # Giá trị này hiện đã được ml_evaluator.py tính tự động
        "utility_score_percent": (
            current_result[
                "utility_score_percent"
            ]
        ),

        # Kiểm tra Strategy có đạt điều kiện privacy
        # k >= 5 và l >= 2 theo tiêu chí của đồ án hay không
        "privacy_valid": (
            current_result[
                "privacy"
            ][
                "privacy_valid"
            ]
        )
    }


    # Thêm Strategy hiện tại
    # vào danh sách dùng để xếp hạng
    strategies_for_selection.append(
        strategy_selection_data
    )


# ============================================================
# HIỂN THỊ DỮ LIỆU ĐẦU VÀO
# ============================================================


# Duyệt từng Strategy để kiểm tra
# dữ liệu trước khi đưa sang Strategy Selector
for strategy_data in strategies_for_selection:

    # Hiển thị tên Strategy
    print(
        "Strategy:",
        strategy_data[
            "strategy"
        ]
    )


    # Hiển thị tỷ lệ suppression
    print(
        "Suppression (%):",
        round(
            strategy_data[
                "suppression_percent"
            ],
            2
        )
    )


    # Hiển thị Utility Score tự động
    print(
        "Validation Utility Score (%):",
        round(
            strategy_data[
                "utility_score_percent"
            ],
            2
        )
    )


    # Hiển thị trạng thái Privacy
    print(
        "Privacy hợp lệ?:",
        strategy_data[
            "privacy_valid"
        ]
    )


    # In dòng phân cách
    print(
        "----------------------------------------"
    )
```

```python
# Import importlib để có thể nạp lại phiên bản mới nhất của strategy_selector.py
import importlib


# Import module chứa các hàm tính điểm và xếp hạng Strategy
import strategy_selector


# Nạp lại module để notebook chắc chắn sử dụng code mới nhất
importlib.reload(
    strategy_selector
)


# Gọi hàm rank_strategies() để tự động xếp hạng các Strategy
# Chỉ các Strategy có privacy_valid=True mới đủ điều kiện được xếp hạng
ranked_strategies = (
    strategy_selector.rank_strategies(

        # Danh sách Strategy đã chuẩn bị ở bước trước
        strategies_for_selection,

        # Trọng số dành cho tỷ lệ dữ liệu được giữ lại
        # 0.5 nghĩa là chiếm 50% Selection Score
        weight_data_retention=0.5,

        # Trọng số dành cho Validation Utility Score
        # 0.5 nghĩa là chiếm 50% Selection Score
        weight_utility=0.5
    )
)


# Lần lượt lấy từng Strategy sau khi đã được xếp hạng
for strategy in ranked_strategies:

    # Hiển thị toàn bộ thông tin của Strategy
    print(
        strategy
    )
```

```python
# ============================================================
# TỰ ĐỘNG CHỌN STRATEGY TỐT NHẤT
# ============================================================


# Gọi hàm select_best_strategy() để hệ thống
# tự chọn Strategy có Selection Score cao nhất
#
# Chỉ những Strategy có privacy_valid=True
# mới đủ điều kiện được lựa chọn
best_strategy = (
    strategy_selector.select_best_strategy(

        # Danh sách các Strategy đã có:
        # - Suppression
        # - Utility
        # - Privacy
        strategies_for_selection,

        # Data Retention chiếm 50% Selection Score
        weight_data_retention=0.5,

        # Validation Utility chiếm 50% Selection Score
        weight_utility=0.5
    )
)


# Hiển thị toàn bộ thông tin
# của Strategy được hệ thống lựa chọn
print(
    best_strategy
)
```

```python
# ============================================================
# LẤY KẾT QUẢ CỦA STRATEGY ĐƯỢC HỆ THỐNG TỰ ĐỘNG CHỌN
# ============================================================


# Lấy tên Strategy tốt nhất
# từ kết quả mà strategy_selector vừa trả về
selected_strategy_name = (
    best_strategy[
        "strategy"
    ]
)


# Lấy dataset ẩn danh cuối cùng
# tương ứng với Strategy đã được chọn
selected_release_data = (
    strategy_results[
        selected_strategy_name
    ][
        "release_data"
    ]
)


# Lấy index gốc của các bản ghi
# được giữ lại trong Strategy đã chọn
selected_retained_indices = (
    strategy_results[
        selected_strategy_name
    ][
        "retained_indices"
    ]
)


# Hiển thị tên Strategy mà hệ thống đã chọn
print(
    "Strategy được chọn:",
    selected_strategy_name
)


# Hiển thị kích thước dataset ẩn danh cuối cùng
print(
    "Kích thước dữ liệu cuối cùng:",
    selected_release_data.shape
)


# Hiển thị số index gốc được giữ lại
print(
    "Số index được giữ lại:",
    len(
        selected_retained_indices
    )
)
```

```python
# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module chứa các hàm đánh giá Machine Learning
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm hàm get_aligned_original_data()
importlib.reload(
    ml_evaluator
)


# Lấy đúng các bản ghi gốc tương ứng với Strategy 1
# strategy1_retained_indices_auto chứa 30.604 index gốc được giữ lại
strategy1_original_aligned = (
    ml_evaluator.get_aligned_original_data(

        # Dataset gốc ban đầu
        data_check,

        # Index gốc còn lại sau khi ẩn danh hóa Strategy 1
        strategy1_retained_indices_auto
    )
)


# Hiển thị kích thước dữ liệu gốc đã được căn chỉnh
# Kỳ vọng phải có 30.604 dòng và vẫn giữ 15 cột gốc
print(
    "Kích thước dữ liệu gốc đã căn chỉnh:",
    strategy1_original_aligned.shape
)


# Hiển thị kích thước dữ liệu ẩn danh Strategy 1
# Kỳ vọng có cùng 30.604 dòng nhưng chỉ còn 8 cột phát hành
print(
    "Kích thước dữ liệu ẩn danh Strategy 1:",
    strategy1_release_auto.shape
)


# Kiểm tra số dòng của dữ liệu gốc và dữ liệu ẩn danh có bằng nhau không
# True nghĩa là hai bộ dữ liệu đang cùng số lượng bản ghi
same_number_of_rows = (
    len(
        strategy1_original_aligned
    )
    ==
    len(
        strategy1_release_auto
    )
)


# Hiển thị kết quả kiểm tra số dòng
print(
    "Hai bộ dữ liệu có cùng số dòng không?:",
    same_number_of_rows
)
```

```python
# ============================================================
# KIỂM TRA DỮ LIỆU GỐC VÀ DỮ LIỆU ẨN DANH
# CÓ THỰC SỰ CĂN CHỈNH ĐÚNG TỪNG BẢN GHI HAY KHÔNG
# ============================================================


# So sánh index của dữ liệu gốc đã căn chỉnh
# với index của dữ liệu ẩn danh Strategy 1
#
# Nếu True nghĩa là hai DataFrame đang chứa
# đúng cùng các bản ghi gốc và cùng thứ tự
same_indices_strategy1 = (
    strategy1_original_aligned.index.tolist()
    ==
    strategy1_release_auto.index.tolist()
)


# So sánh cột thu_nhap của hai bộ dữ liệu
# Vì thu_nhap không bị thay đổi trong quá trình ẩn danh hóa
# nên từng giá trị thu_nhap phải giống nhau hoàn toàn
same_target_strategy1 = (
    strategy1_original_aligned["thu_nhap"].tolist()
    ==
    strategy1_release_auto["thu_nhap"].tolist()
)


# Hiển thị kết quả kiểm tra index
print(
    "Index hai bộ dữ liệu có giống hoàn toàn không?:",
    same_indices_strategy1
)


# Hiển thị kết quả kiểm tra target thu_nhap
print(
    "Target thu_nhap có giống hoàn toàn không?:",
    same_target_strategy1
)
```

```python
# ============================================================
# KIỂM TRA HÀM CHIA TRAIN / VALIDATION / TEST
# VỚI STRATEGY 1
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module chứa các hàm đánh giá Machine Learning
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm split_train_validation_test_indices()
importlib.reload(
    ml_evaluator
)


# Chia 30.604 bản ghi của Strategy 1 thành:
# - Train      = 60%
# - Validation = 20%
# - Test       = 20%
#
# Hàm chỉ chia INDEX để sau này dữ liệu gốc
# và dữ liệu ẩn danh sử dụng đúng cùng các bản ghi
train_indices_s1, validation_indices_s1, test_indices_s1 = (
    ml_evaluator.split_train_validation_test_indices(

        # Dữ liệu gốc đã được căn chỉnh đúng với Strategy 1
        strategy1_original_aligned,

        # Cột thu_nhap là target dùng để stratify
        target_column="thu_nhap",

        # Cố định random_state để kết quả có thể tái lập
        random_state=42
    )
)


# Hiển thị số bản ghi trong tập Train
print(
    "Số bản ghi Train:",
    len(
        train_indices_s1
    )
)


# Hiển thị số bản ghi trong tập Validation
print(
    "Số bản ghi Validation:",
    len(
        validation_indices_s1
    )
)


# Hiển thị số bản ghi trong tập Test
print(
    "Số bản ghi Test:",
    len(
        test_indices_s1
    )
)


# Tính tổng số bản ghi sau khi chia
# Tổng này phải bằng đúng 30.604 bản ghi ban đầu
total_split_records_s1 = (
    len(train_indices_s1)
    + len(validation_indices_s1)
    + len(test_indices_s1)
)


# Hiển thị tổng số bản ghi sau khi chia
print(
    "Tổng số bản ghi sau khi chia:",
    total_split_records_s1
)
```

```python
# ============================================================
# KIỂM TRA ORIGINAL VÀ ANONYMIZED
# CÓ ĐƯỢC CHIA ĐÚNG CÙNG TRAIN / VALIDATION / TEST KHÔNG
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module chứa các hàm đánh giá Machine Learning
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm hàm get_data_splits_by_indices()
importlib.reload(
    ml_evaluator
)


# ============================================================
# CHIA DỮ LIỆU GỐC STRATEGY 1
# ============================================================


# Lấy Train / Validation / Test của dữ liệu gốc
# bằng đúng 3 bộ index đã chia ở bước trước
original_train_s1, original_validation_s1, original_test_s1 = (
    ml_evaluator.get_data_splits_by_indices(

        # Dữ liệu gốc đã căn chỉnh còn đúng 30.604 bản ghi
        strategy1_original_aligned,

        # Index của tập Train
        train_indices_s1,

        # Index của tập Validation
        validation_indices_s1,

        # Index của tập Test
        test_indices_s1
    )
)


# ============================================================
# CHIA DỮ LIỆU ẨN DANH STRATEGY 1
# ============================================================


# Lấy Train / Validation / Test của dữ liệu ẩn danh
# bằng CHÍNH XÁC cùng 3 bộ index ở trên
anonymized_train_s1, anonymized_validation_s1, anonymized_test_s1 = (
    ml_evaluator.get_data_splits_by_indices(

        # Dataset ẩn danh Strategy 1
        strategy1_release_auto,

        # Dùng cùng index Train với dữ liệu gốc
        train_indices_s1,

        # Dùng cùng index Validation với dữ liệu gốc
        validation_indices_s1,

        # Dùng cùng index Test với dữ liệu gốc
        test_indices_s1
    )
)


# ============================================================
# KIỂM TRA KÍCH THƯỚC
# ============================================================


# Hiển thị kích thước Train của Original và Anonymized
print(
    "Original Train:",
    original_train_s1.shape,
    "| Anonymized Train:",
    anonymized_train_s1.shape
)


# Hiển thị kích thước Validation của Original và Anonymized
print(
    "Original Validation:",
    original_validation_s1.shape,
    "| Anonymized Validation:",
    anonymized_validation_s1.shape
)


# Hiển thị kích thước Test của Original và Anonymized
print(
    "Original Test:",
    original_test_s1.shape,
    "| Anonymized Test:",
    anonymized_test_s1.shape
)


# ============================================================
# KIỂM TRA TARGET THU_NHAP
# ============================================================


# Kiểm tra target của tập Train có giống hoàn toàn không
train_target_match = (
    original_train_s1["thu_nhap"].tolist()
    ==
    anonymized_train_s1["thu_nhap"].tolist()
)


# Kiểm tra target của tập Validation có giống hoàn toàn không
validation_target_match = (
    original_validation_s1["thu_nhap"].tolist()
    ==
    anonymized_validation_s1["thu_nhap"].tolist()
)


# Kiểm tra target của tập Test có giống hoàn toàn không
test_target_match = (
    original_test_s1["thu_nhap"].tolist()
    ==
    anonymized_test_s1["thu_nhap"].tolist()
)


# Hiển thị kết quả kiểm tra target Train
print(
    "Target Train giống nhau?:",
    train_target_match
)


# Hiển thị kết quả kiểm tra target Validation
print(
    "Target Validation giống nhau?:",
    validation_target_match
)


# Hiển thị kết quả kiểm tra target Test
print(
    "Target Test giống nhau?:",
    test_target_match
)
```

```python
# ============================================================
# KIỂM TRA HÀM CHUẨN BỊ FEATURE VÀ TARGET CỦA ORIGINAL
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module chứa các hàm đánh giá Machine Learning
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm prepare_original_features_target()
importlib.reload(
    ml_evaluator
)


# Gọi hàm để tách:
# - X = 7 feature gốc dùng cho Machine Learning
# - y = target thu_nhap
#
# Dữ liệu đầu vào là tập Train của Original Strategy 1
X_original_train_s1, y_original_train_s1 = (
    ml_evaluator.prepare_original_features_target(
        original_train_s1,
        target_column="thu_nhap"
    )
)


# Hiển thị kích thước của X
# Kỳ vọng có 18.362 dòng và 7 feature
print(
    "Kích thước X Original Train:",
    X_original_train_s1.shape
)


# Hiển thị số lượng giá trị target
# Kỳ vọng có đúng 18.362 nhãn thu_nhap
print(
    "Số lượng y Original Train:",
    len(
        y_original_train_s1
    )
)


# Hiển thị tên 7 feature
# để chắc chắn không lấy nhầm các cột khác của dataset gốc
print(
    "Các feature Original:",
    X_original_train_s1.columns.tolist()
)


# Đếm số missing value sau khi dấu ? đã được chuyển thành np.nan
# Mục đích là xác nhận dữ liệu thiếu đã được chuẩn bị đúng
print(
    "Số missing value trong từng feature:"
)


# Hiển thị số lượng missing của từng cột
print(
    X_original_train_s1
    .isna()
    .sum()
)
```

```python
# ============================================================
# KIỂM TRA HÀM CHUẨN BỊ FEATURE VÀ TARGET
# CỦA DỮ LIỆU ẨN DANH STRATEGY 1
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module chứa các hàm đánh giá Machine Learning
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm prepare_anonymized_features_target()
importlib.reload(
    ml_evaluator
)


# Tách dữ liệu ẩn danh Train của Strategy 1 thành:
# - X = 7 feature đã được ẩn danh hóa
# - y = target thu_nhap
X_anonymized_train_s1, y_anonymized_train_s1 = (
    ml_evaluator.prepare_anonymized_features_target(
        anonymized_train_s1,
        target_column="thu_nhap"
    )
)


# Hiển thị kích thước X của dữ liệu ẩn danh
# Kỳ vọng: 18.362 dòng và 7 feature
print(
    "Kích thước X Anonymized Train:",
    X_anonymized_train_s1.shape
)


# Hiển thị số lượng target
# Kỳ vọng: 18.362 giá trị
print(
    "Số lượng y Anonymized Train:",
    len(
        y_anonymized_train_s1
    )
)


# Hiển thị tên các feature sau khi ẩn danh hóa
print(
    "Các feature Anonymized:",
    X_anonymized_train_s1.columns.tolist()
)


# Kiểm tra target Original và Anonymized có giống nhau không
# True nghĩa là hai mô hình sẽ học trên đúng cùng nhãn
same_train_target = (
    y_original_train_s1.tolist()
    ==
    y_anonymized_train_s1.tolist()
)


# Hiển thị kết quả kiểm tra target
print(
    "Target Original và Anonymized giống nhau?:",
    same_train_target
)
```

```python
# ============================================================
# KIỂM TRA BỘ PREPROCESSING CỦA DỮ LIỆU ORIGINAL
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module ml_evaluator
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm build_original_preprocessor()
importlib.reload(
    ml_evaluator
)


# Tạo bộ preprocessing dành cho dữ liệu Original
# Lúc này preprocessor mới chỉ được tạo, chưa học từ dữ liệu
original_preprocessor_s1 = (
    ml_evaluator.build_original_preprocessor()
)


# Fit preprocessing CHỈ trên tập Train
# Điều này rất quan trọng để tránh data leakage
#
# fit_transform() sẽ:
# - học median của các cột số
# - học giá trị phổ biến nhất của cột phân loại
# - học các category cho OneHotEncoder
# - chuẩn hóa các cột số
# - chuyển toàn bộ dữ liệu thành dạng số
X_original_train_processed_s1 = (
    original_preprocessor_s1.fit_transform(
        X_original_train_s1
    )
)


# Hiển thị kích thước dữ liệu sau preprocessing
# Số dòng vẫn phải là 18.362
# Số cột sẽ tăng vì OneHotEncoder tạo thêm nhiều cột 0/1
print(
    "Kích thước Original Train sau preprocessing:",
    X_original_train_processed_s1.shape
)


# Hiển thị kiểu dữ liệu sau preprocessing
# Kết quả có thể là numpy array hoặc sparse matrix
print(
    "Kiểu dữ liệu sau preprocessing:",
    type(
        X_original_train_processed_s1
    )
)
```

```python
# ============================================================
# KIỂM TRA BỘ PREPROCESSING CỦA DỮ LIỆU ANONYMIZED
# STRATEGY 1
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module ml_evaluator
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm build_anonymized_preprocessor()
importlib.reload(
    ml_evaluator
)


# Lấy danh sách tên 7 feature của dữ liệu ẩn danh Strategy 1
# Danh sách này sẽ được truyền vào bộ preprocessing
anonymized_feature_columns_s1 = (
    X_anonymized_train_s1.columns.tolist()
)


# Tạo bộ preprocessing dành cho dữ liệu ẩn danh Strategy 1
# Lúc này preprocessor mới chỉ được tạo, chưa học từ dữ liệu
anonymized_preprocessor_s1 = (
    ml_evaluator.build_anonymized_preprocessor(
        anonymized_feature_columns_s1
    )
)


# Fit preprocessing CHỈ trên tập Train
# để tránh data leakage từ Validation hoặc Test
#
# fit_transform() sẽ:
# - học giá trị phổ biến nhất nếu có missing
# - học các category xuất hiện trong Train
# - chuyển category thành các cột 0/1 bằng OneHotEncoder
X_anonymized_train_processed_s1 = (
    anonymized_preprocessor_s1.fit_transform(
        X_anonymized_train_s1
    )
)


# Hiển thị kích thước dữ liệu ẩn danh sau preprocessing
# Số dòng phải giữ nguyên là 18.362
print(
    "Kích thước Anonymized Train sau preprocessing:",
    X_anonymized_train_processed_s1.shape
)


# Hiển thị kiểu dữ liệu sau preprocessing
# Thường sẽ là sparse matrix do OneHotEncoder tạo ra
print(
    "Kiểu dữ liệu sau preprocessing:",
    type(
        X_anonymized_train_processed_s1
    )
)
```

```python
# ============================================================
# KIỂM TRA HÀM TẠO PIPELINE LOGISTIC REGRESSION
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất của ml_evaluator.py
import importlib


# Import module ml_evaluator
import ml_evaluator


# Nạp lại module vì chúng ta vừa thêm
# hàm build_logistic_regression_pipeline()
importlib.reload(
    ml_evaluator
)


# Tạo một bộ preprocessing mới dành cho dữ liệu Original
# Chúng ta tạo mới để pipeline chưa bị ảnh hưởng
# bởi lần fit thử preprocessing trước đó
original_preprocessor_for_model_s1 = (
    ml_evaluator.build_original_preprocessor()
)


# Tạo Pipeline hoàn chỉnh gồm:
# preprocessing → Logistic Regression
original_model_pipeline_s1 = (
    ml_evaluator.build_logistic_regression_pipeline(
        original_preprocessor_for_model_s1
    )
)


# Hiển thị kiểu đối tượng được tạo
# Kỳ vọng đây phải là sklearn Pipeline
print(
    "Kiểu pipeline:",
    type(
        original_model_pipeline_s1
    )
)


# Hiển thị tên các bước có trong Pipeline
# Kỳ vọng gồm đúng:
# preprocessor và model
print(
    "Các bước trong pipeline:",
    list(
        original_model_pipeline_s1.named_steps.keys()
    )
)


# Kiểm tra loại mô hình nằm ở bước model
# Kỳ vọng là LogisticRegression
print(
    "Mô hình:",
    type(
        original_model_pipeline_s1.named_steps["model"]
    )
)


# Kiểm tra max_iter đã được cấu hình đúng là 1000 chưa
print(
    "max_iter:",
    original_model_pipeline_s1.named_steps["model"].max_iter
)
```

```python
# ============================================================
# HUẤN LUYỆN LOGISTIC REGRESSION
# TRÊN DỮ LIỆU ORIGINAL TRAIN CỦA STRATEGY 1
# ============================================================


# Huấn luyện toàn bộ Pipeline trên tập Train
#
# Pipeline sẽ tự động thực hiện:
# 1. Điền missing value
# 2. Chuẩn hóa các feature số
# 3. One-Hot Encoding các feature phân loại
# 4. Huấn luyện Logistic Regression
original_model_pipeline_s1.fit(
    X_original_train_s1,
    y_original_train_s1
)


# Lấy mô hình Logistic Regression
# sau khi đã được huấn luyện xong
trained_original_model_s1 = (
    original_model_pipeline_s1
    .named_steps[
        "model"
    ]
)


# Hiển thị các lớp mà mô hình đã học
# Kỳ vọng gồm hai lớp <=50K và >50K
print(
    "Các lớp mô hình đã học:",
    trained_original_model_s1.classes_
)


# Hiển thị kích thước hệ số của Logistic Regression
# Nếu preprocessing đúng thì mô hình phải nhận 71 feature
print(
    "Kích thước hệ số mô hình:",
    trained_original_model_s1.coef_.shape
)
```

```python
# ============================================================
# HUẤN LUYỆN LOGISTIC REGRESSION
# TRÊN DỮ LIỆU ANONYMIZED TRAIN CỦA STRATEGY 1
# ============================================================


# Lấy danh sách 7 feature của dữ liệu ẩn danh Strategy 1
# Danh sách này sẽ được dùng để tạo bộ preprocessing
anonymized_feature_columns_for_model_s1 = (
    X_anonymized_train_s1.columns.tolist()
)


# Tạo một bộ preprocessing MỚI cho dữ liệu ẩn danh
# Không dùng lại preprocessor đã fit thử trước đó
anonymized_preprocessor_for_model_s1 = (
    ml_evaluator.build_anonymized_preprocessor(
        anonymized_feature_columns_for_model_s1
    )
)


# Tạo Pipeline gồm:
# preprocessing → Logistic Regression
anonymized_model_pipeline_s1 = (
    ml_evaluator.build_logistic_regression_pipeline(
        anonymized_preprocessor_for_model_s1
    )
)


# Huấn luyện Pipeline trên tập Anonymized Train
# Pipeline sẽ tự:
# - xử lý missing nếu có
# - One-Hot Encoding
# - huấn luyện Logistic Regression
anonymized_model_pipeline_s1.fit(
    X_anonymized_train_s1,
    y_anonymized_train_s1
)


# Lấy mô hình Logistic Regression
# sau khi đã huấn luyện xong
trained_anonymized_model_s1 = (
    anonymized_model_pipeline_s1
    .named_steps[
        "model"
    ]
)


# Hiển thị các lớp mà mô hình đã học
print(
    "Các lớp mô hình Anonymized đã học:",
    trained_anonymized_model_s1.classes_
)


# Hiển thị kích thước hệ số của mô hình
# Kỳ vọng mô hình nhận 21 feature sau preprocessing
print(
    "Kích thước hệ số mô hình Anonymized:",
    trained_anonymized_model_s1.coef_.shape
)
```

```python
# ============================================================
# CHUẨN BỊ DỮ LIỆU VALIDATION CHO STRATEGY 1
# ============================================================


# Chuẩn bị X và y của dữ liệu Original Validation
# Hàm sẽ lấy đúng 7 feature gốc
# và chuyển dấu ? thành missing value nếu có
X_original_validation_s1, y_original_validation_s1 = (
    ml_evaluator.prepare_original_features_target(
        original_validation_s1,
        target_column="thu_nhap"
    )
)


# Chuẩn bị X và y của dữ liệu Anonymized Validation
# Hàm sẽ lấy 7 feature đã được generalization
# và tách riêng target thu_nhap
X_anonymized_validation_s1, y_anonymized_validation_s1 = (
    ml_evaluator.prepare_anonymized_features_target(
        anonymized_validation_s1,
        target_column="thu_nhap"
    )
)


# Hiển thị kích thước X Original Validation
# Kỳ vọng có 6.121 dòng và 7 feature
print(
    "X Original Validation:",
    X_original_validation_s1.shape
)


# Hiển thị kích thước X Anonymized Validation
# Kỳ vọng cũng có 6.121 dòng và 7 feature
print(
    "X Anonymized Validation:",
    X_anonymized_validation_s1.shape
)


# Kiểm tra số lượng target của Original
print(
    "Số lượng y Original Validation:",
    len(
        y_original_validation_s1
    )
)


# Kiểm tra số lượng target của Anonymized
print(
    "Số lượng y Anonymized Validation:",
    len(
        y_anonymized_validation_s1
    )
)


# So sánh target của hai tập Validation
# True nghĩa là Original và Anonymized
# đang được đánh giá trên chính xác cùng nhãn
same_validation_target_s1 = (
    y_original_validation_s1.tolist()
    ==
    y_anonymized_validation_s1.tolist()
)


# Hiển thị kết quả kiểm tra target
print(
    "Target Validation giống nhau?:",
    same_validation_target_s1
)
```

```python
# ============================================================
# DỰ ĐOÁN TRÊN TẬP VALIDATION
# CHO ORIGINAL VÀ ANONYMIZED STRATEGY 1
# ============================================================


# ============================================================
# 1. DỰ ĐOÁN VỚI MÔ HÌNH ORIGINAL
# ============================================================


# Dùng mô hình Original đã huấn luyện
# để dự đoán nhãn thu_nhap trên tập Validation
y_pred_original_validation_s1 = (
    original_model_pipeline_s1.predict(
        X_original_validation_s1
    )
)


# Lấy danh sách các lớp mà mô hình Original đã học
# Ví dụ: ['<=50K', '>50K']
original_classes_s1 = (
    original_model_pipeline_s1
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
# Không viết cứng [:, 1] để code an toàn hơn
original_positive_index_s1 = (
    list(
        original_classes_s1
    )
    .index(
        ">50K"
    )
)


# Dự đoán xác suất của lớp >50K
# Xác suất này sẽ được dùng để tính ROC-AUC
y_prob_original_validation_s1 = (
    original_model_pipeline_s1.predict_proba(
        X_original_validation_s1
    )[
        :,
        original_positive_index_s1
    ]
)


# ============================================================
# 2. DỰ ĐOÁN VỚI MÔ HÌNH ANONYMIZED
# ============================================================


# Dùng mô hình Anonymized đã huấn luyện
# để dự đoán nhãn trên cùng tập Validation
y_pred_anonymized_validation_s1 = (
    anonymized_model_pipeline_s1.predict(
        X_anonymized_validation_s1
    )
)


# Lấy danh sách các lớp mà mô hình Anonymized đã học
anonymized_classes_s1 = (
    anonymized_model_pipeline_s1
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
anonymized_positive_index_s1 = (
    list(
        anonymized_classes_s1
    )
    .index(
        ">50K"
    )
)


# Dự đoán xác suất thuộc lớp >50K
# để sau này tính ROC-AUC cho dữ liệu ẩn danh
y_prob_anonymized_validation_s1 = (
    anonymized_model_pipeline_s1.predict_proba(
        X_anonymized_validation_s1
    )[
        :,
        anonymized_positive_index_s1
    ]
)


# ============================================================
# 3. KIỂM TRA KẾT QUẢ DỰ ĐOÁN
# ============================================================


# Kiểm tra số lượng nhãn dự đoán Original
# Phải bằng đúng 6.121 bản ghi Validation
print(
    "Số dự đoán Original:",
    len(
        y_pred_original_validation_s1
    )
)


# Kiểm tra số lượng xác suất Original
# Cũng phải bằng 6.121
print(
    "Số xác suất Original:",
    len(
        y_prob_original_validation_s1
    )
)


# Kiểm tra số lượng nhãn dự đoán Anonymized
print(
    "Số dự đoán Anonymized:",
    len(
        y_pred_anonymized_validation_s1
    )
)


# Kiểm tra số lượng xác suất Anonymized
print(
    "Số xác suất Anonymized:",
    len(
        y_prob_anonymized_validation_s1
    )
)


# Kiểm tra xác suất nhỏ nhất và lớn nhất của Original
# Các giá trị phải nằm trong khoảng từ 0 đến 1
print(
    "Khoảng xác suất Original:",
    y_prob_original_validation_s1.min(),
    "→",
    y_prob_original_validation_s1.max()
)


# Kiểm tra xác suất nhỏ nhất và lớn nhất của Anonymized
# Các giá trị cũng phải nằm trong khoảng từ 0 đến 1
print(
    "Khoảng xác suất Anonymized:",
    y_prob_anonymized_validation_s1.min(),
    "→",
    y_prob_anonymized_validation_s1.max()
)
```

```python
# ============================================================
# TÍNH CÁC METRIC VALIDATION
# CHO ORIGINAL VÀ ANONYMIZED STRATEGY 1
# ============================================================


# Import importlib để có thể nạp lại module mới nhất
import importlib


# Import module utility_evaluator
# Module này chứa hàm tính Accuracy, Precision, Recall,
# F1-score, Balanced Accuracy và ROC-AUC
import utility_evaluator


# Nạp lại utility_evaluator để chắc chắn
# notebook đang dùng phiên bản code mới nhất
importlib.reload(
    utility_evaluator
)


# ============================================================
# 1. TÍNH METRIC CHO ORIGINAL
# ============================================================


# Tính các chỉ số đánh giá cho mô hình Original
original_validation_metrics_s1 = (
    utility_evaluator.evaluate_classification_metrics(

        # Nhãn thật của tập Original Validation
        y_true=y_original_validation_s1,

        # Nhãn mà mô hình Original dự đoán
        y_pred=y_pred_original_validation_s1,

        # Xác suất dự đoán lớp >50K
        y_prob=y_prob_original_validation_s1,

        # Xác định >50K là lớp dương
        positive_label=">50K"
    )
)


# ============================================================
# 2. TÍNH METRIC CHO ANONYMIZED
# ============================================================


# Tính các chỉ số đánh giá cho mô hình Anonymized
anonymized_validation_metrics_s1 = (
    utility_evaluator.evaluate_classification_metrics(

        # Nhãn thật của tập Anonymized Validation
        y_true=y_anonymized_validation_s1,

        # Nhãn mà mô hình Anonymized dự đoán
        y_pred=y_pred_anonymized_validation_s1,

        # Xác suất dự đoán lớp >50K
        y_prob=y_prob_anonymized_validation_s1,

        # Xác định >50K là lớp dương
        positive_label=">50K"
    )
)


# ============================================================
# 3. HIỂN THỊ KẾT QUẢ ORIGINAL
# ============================================================


# In tiêu đề kết quả của Original
print(
    "===== ORIGINAL VALIDATION ====="
)


# Duyệt từng metric của Original
for metric_name, metric_value in original_validation_metrics_s1.items():

    # Hiển thị tên metric và giá trị dưới dạng phần trăm
    print(
        metric_name,
        ":",
        round(
            metric_value * 100,
            2
        ),
        "%"
    )


# ============================================================
# 4. HIỂN THỊ KẾT QUẢ ANONYMIZED
# ============================================================


# In tiêu đề kết quả của Anonymized
print(
    "\n===== ANONYMIZED VALIDATION ====="
)


# Duyệt từng metric của Anonymized
for metric_name, metric_value in anonymized_validation_metrics_s1.items():

    # Hiển thị tên metric và giá trị dưới dạng phần trăm
    print(
        metric_name,
        ":",
        round(
            metric_value * 100,
            2
        ),
        "%"
    )
```

```python
# ============================================================
# TẠO LẠI TRAIN / VALIDATION / TEST
# SAU KHI ĐÃ SỬA CÁCH CHIA INDEX
# ============================================================


# Lấy lại 3 tập của dữ liệu Original
# bằng bộ index mới vừa được xác nhận là đúng
original_train_s1, original_validation_s1, original_test_s1 = (
    ml_evaluator.get_data_splits_by_indices(

        # Dữ liệu gốc đã căn chỉnh với Strategy 1
        strategy1_original_aligned,

        # Index Train mới
        train_indices_s1,

        # Index Validation mới
        validation_indices_s1,

        # Index Test mới
        test_indices_s1
    )
)


# Lấy lại 3 tập của dữ liệu Anonymized
# bằng CHÍNH XÁC cùng bộ index với Original
anonymized_train_s1, anonymized_validation_s1, anonymized_test_s1 = (
    ml_evaluator.get_data_splits_by_indices(

        # Dữ liệu ẩn danh của Strategy 1
        strategy1_release_auto,

        # Dùng cùng index Train
        train_indices_s1,

        # Dùng cùng index Validation
        validation_indices_s1,

        # Dùng cùng index Test
        test_indices_s1
    )
)


# Kiểm tra kích thước tập Train
print(
    "Train:",
    original_train_s1.shape,
    anonymized_train_s1.shape
)


# Kiểm tra kích thước tập Validation
print(
    "Validation:",
    original_validation_s1.shape,
    anonymized_validation_s1.shape
)


# Kiểm tra kích thước tập Test
print(
    "Test:",
    original_test_s1.shape,
    anonymized_test_s1.shape
)


# Kiểm tra target Train giữa Original và Anonymized
# True nghĩa là hai bên đang chứa đúng cùng nhãn
print(
    "Target Train giống nhau?:",
    original_train_s1["thu_nhap"].tolist()
    ==
    anonymized_train_s1["thu_nhap"].tolist()
)


# Kiểm tra target Validation giữa Original và Anonymized
print(
    "Target Validation giống nhau?:",
    original_validation_s1["thu_nhap"].tolist()
    ==
    anonymized_validation_s1["thu_nhap"].tolist()
)


# Kiểm tra target Test giữa Original và Anonymized
print(
    "Target Test giống nhau?:",
    original_test_s1["thu_nhap"].tolist()
    ==
    anonymized_test_s1["thu_nhap"].tolist()
)
```

```python
# ============================================================
# TẠO LẠI X VÀ y CỦA TẬP TRAIN
# SAU KHI ĐÃ SỬA CÁCH CHIA DỮ LIỆU
# ============================================================


# Tạo lại feature X và target y
# cho dữ liệu Original Train mới
X_original_train_s1, y_original_train_s1 = (
    ml_evaluator.prepare_original_features_target(

        # Tập Original Train vừa được tạo lại bằng index đúng
        original_train_s1,

        # Cột cần dự đoán của bài toán Machine Learning
        target_column="thu_nhap"
    )
)


# Tạo lại feature X và target y
# cho dữ liệu Anonymized Train mới
X_anonymized_train_s1, y_anonymized_train_s1 = (
    ml_evaluator.prepare_anonymized_features_target(

        # Tập Anonymized Train vừa được tạo lại bằng cùng index
        anonymized_train_s1,

        # Cột target không bị thay đổi khi ẩn danh hóa
        target_column="thu_nhap"
    )
)


# Kiểm tra kích thước X Original
# Phải có 18.362 dòng và 7 feature
print(
    "X Original Train:",
    X_original_train_s1.shape
)


# Kiểm tra kích thước X Anonymized
# Cũng phải có 18.362 dòng và 7 feature
print(
    "X Anonymized Train:",
    X_anonymized_train_s1.shape
)


# Kiểm tra số lượng target Original
print(
    "y Original Train:",
    len(
        y_original_train_s1
    )
)


# Kiểm tra số lượng target Anonymized
print(
    "y Anonymized Train:",
    len(
        y_anonymized_train_s1
    )
)


# So sánh target của hai tập Train
# True nghĩa là hai mô hình sẽ được huấn luyện
# trên chính xác cùng 18.362 nhãn
print(
    "Target Train giống nhau?:",
    y_original_train_s1.tolist()
    ==
    y_anonymized_train_s1.tolist()
)
```

```python
# ============================================================
# TẠO MỚI VÀ HUẤN LUYỆN LẠI HAI MÔ HÌNH
# SAU KHI ĐÃ SỬA CÁCH CHIA DỮ LIỆU
# ============================================================


# ============================================================
# 1. TẠO VÀ HUẤN LUYỆN MÔ HÌNH ORIGINAL
# ============================================================


# Tạo một bộ preprocessing MỚI cho dữ liệu Original
# Không dùng lại preprocessor đã fit trên cách chia cũ
original_preprocessor_s1 = (
    ml_evaluator.build_original_preprocessor()
)


# Tạo một Pipeline Logistic Regression MỚI
# gồm preprocessing và mô hình Logistic Regression
original_model_pipeline_s1 = (
    ml_evaluator.build_logistic_regression_pipeline(
        original_preprocessor_s1
    )
)


# Huấn luyện mô hình Original
# trên đúng 18.362 bản ghi Train mới
original_model_pipeline_s1.fit(
    X_original_train_s1,
    y_original_train_s1
)


# ============================================================
# 2. TẠO VÀ HUẤN LUYỆN MÔ HÌNH ANONYMIZED
# ============================================================


# Lấy danh sách 7 feature của Strategy 1
# để tạo preprocessing cho dữ liệu ẩn danh
anonymized_feature_columns_s1 = (
    X_anonymized_train_s1
    .columns
    .tolist()
)


# Tạo một bộ preprocessing MỚI
# cho dữ liệu Anonymized Strategy 1
anonymized_preprocessor_s1 = (
    ml_evaluator.build_anonymized_preprocessor(
        anonymized_feature_columns_s1
    )
)


# Tạo một Pipeline Logistic Regression MỚI
# dành cho dữ liệu Anonymized
anonymized_model_pipeline_s1 = (
    ml_evaluator.build_logistic_regression_pipeline(
        anonymized_preprocessor_s1
    )
)


# Huấn luyện mô hình Anonymized
# trên đúng cùng 18.362 bản ghi Train
anonymized_model_pipeline_s1.fit(
    X_anonymized_train_s1,
    y_anonymized_train_s1
)


# ============================================================
# 3. KIỂM TRA SAU KHI HUẤN LUYỆN
# ============================================================


# Lấy số feature mà mô hình Original thực sự nhận
# sau preprocessing
original_coef_shape_s1 = (
    original_model_pipeline_s1
    .named_steps["model"]
    .coef_
    .shape
)


# Lấy số feature mà mô hình Anonymized thực sự nhận
# sau preprocessing
anonymized_coef_shape_s1 = (
    anonymized_model_pipeline_s1
    .named_steps["model"]
    .coef_
    .shape
)


# Hiển thị kích thước hệ số của mô hình Original
print(
    "Original model coef:",
    original_coef_shape_s1
)


# Hiển thị kích thước hệ số của mô hình Anonymized
print(
    "Anonymized model coef:",
    anonymized_coef_shape_s1
)


# Xác nhận hai mô hình đều đã được huấn luyện
print(
    "Đã huấn luyện lại hai mô hình thành công."
)
```

```python
# ============================================================
# TẠO LẠI X VÀ y CỦA TẬP VALIDATION
# SAU KHI ĐÃ SỬA CÁCH CHIA DỮ LIỆU
# ============================================================


# Chuẩn bị lại feature X và target y
# cho tập Original Validation mới
X_original_validation_s1, y_original_validation_s1 = (
    ml_evaluator.prepare_original_features_target(

        # Tập Original Validation được tạo từ index đúng
        original_validation_s1,

        # Cột target cần dự đoán
        target_column="thu_nhap"
    )
)


# Chuẩn bị lại feature X và target y
# cho tập Anonymized Validation mới
X_anonymized_validation_s1, y_anonymized_validation_s1 = (
    ml_evaluator.prepare_anonymized_features_target(

        # Tập Anonymized Validation dùng cùng index với Original
        anonymized_validation_s1,

        # Cột target không bị thay đổi khi ẩn danh hóa
        target_column="thu_nhap"
    )
)


# Kiểm tra kích thước X Original Validation
# Kỳ vọng có 6.121 dòng và 7 feature
print(
    "X Original Validation:",
    X_original_validation_s1.shape
)


# Kiểm tra kích thước X Anonymized Validation
# Kỳ vọng cũng có 6.121 dòng và 7 feature
print(
    "X Anonymized Validation:",
    X_anonymized_validation_s1.shape
)


# Kiểm tra số lượng target Original Validation
print(
    "y Original Validation:",
    len(
        y_original_validation_s1
    )
)


# Kiểm tra số lượng target Anonymized Validation
print(
    "y Anonymized Validation:",
    len(
        y_anonymized_validation_s1
    )
)


# Kiểm tra target của Original và Anonymized
# có giống nhau hoàn toàn hay không
print(
    "Target Validation giống nhau?:",
    y_original_validation_s1.tolist()
    ==
    y_anonymized_validation_s1.tolist()
)
```

```python
# ============================================================
# DỰ ĐOÁN LẠI TRÊN VALIDATION
# SAU KHI ĐÃ SỬA CÁCH CHIA DỮ LIỆU
# ============================================================


# ============================================================
# 1. DỰ ĐOÁN VỚI MÔ HÌNH ORIGINAL
# ============================================================


# Dùng mô hình Original vừa huấn luyện lại
# để dự đoán nhãn thu_nhap trên tập Validation mới
y_pred_original_validation_s1 = (
    original_model_pipeline_s1.predict(
        X_original_validation_s1
    )
)


# Lấy danh sách các lớp mà mô hình Original đã học
# Kỳ vọng gồm <=50K và >50K
original_classes_s1 = (
    original_model_pipeline_s1
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
# để lấy đúng xác suất dùng cho ROC-AUC
original_positive_index_s1 = (
    list(
        original_classes_s1
    )
    .index(
        ">50K"
    )
)


# Lấy xác suất dự đoán thuộc lớp >50K
# cho từng bản ghi Validation
y_prob_original_validation_s1 = (
    original_model_pipeline_s1.predict_proba(
        X_original_validation_s1
    )[
        :,
        original_positive_index_s1
    ]
)


# ============================================================
# 2. DỰ ĐOÁN VỚI MÔ HÌNH ANONYMIZED
# ============================================================


# Dùng mô hình Anonymized vừa huấn luyện lại
# để dự đoán trên tập Validation tương ứng
y_pred_anonymized_validation_s1 = (
    anonymized_model_pipeline_s1.predict(
        X_anonymized_validation_s1
    )
)


# Lấy danh sách các lớp mà mô hình Anonymized đã học
anonymized_classes_s1 = (
    anonymized_model_pipeline_s1
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
# trong mô hình Anonymized
anonymized_positive_index_s1 = (
    list(
        anonymized_classes_s1
    )
    .index(
        ">50K"
    )
)


# Lấy xác suất dự đoán thuộc lớp >50K
# để sau này tính ROC-AUC
y_prob_anonymized_validation_s1 = (
    anonymized_model_pipeline_s1.predict_proba(
        X_anonymized_validation_s1
    )[
        :,
        anonymized_positive_index_s1
    ]
)


# ============================================================
# 3. KIỂM TRA SỐ LƯỢNG DỰ ĐOÁN
# ============================================================


# Kiểm tra số nhãn dự đoán của Original
print(
    "Số dự đoán Original:",
    len(
        y_pred_original_validation_s1
    )
)


# Kiểm tra số xác suất của Original
print(
    "Số xác suất Original:",
    len(
        y_prob_original_validation_s1
    )
)


# Kiểm tra số nhãn dự đoán của Anonymized
print(
    "Số dự đoán Anonymized:",
    len(
        y_pred_anonymized_validation_s1
    )
)


# Kiểm tra số xác suất của Anonymized
print(
    "Số xác suất Anonymized:",
    len(
        y_prob_anonymized_validation_s1
    )
)
```

```python
# ============================================================
# TÍNH LẠI METRIC VALIDATION CHO STRATEGY 1
# SAU KHI ĐÃ SỬA ĐÚNG CÁCH CHIA DỮ LIỆU
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất
# của module utility_evaluator
import importlib


# Import module chứa các hàm tính metric Machine Learning
import utility_evaluator


# Nạp lại module để chắc chắn notebook
# đang sử dụng phiên bản code mới nhất
importlib.reload(
    utility_evaluator
)


# ============================================================
# 1. TÍNH METRIC CHO ORIGINAL
# ============================================================


# Tính Accuracy, Precision, Recall, F1-score,
# Balanced Accuracy và ROC-AUC của mô hình Original
original_validation_metrics_s1 = (
    utility_evaluator.evaluate_classification_metrics(

        # Nhãn thật của tập Validation
        y_true=y_original_validation_s1,

        # Nhãn dự đoán của mô hình Original
        y_pred=y_pred_original_validation_s1,

        # Xác suất dự đoán lớp >50K
        y_prob=y_prob_original_validation_s1,

        # Xác định >50K là lớp dương
        positive_label=">50K"
    )
)


# ============================================================
# 2. TÍNH METRIC CHO ANONYMIZED
# ============================================================


# Tính các metric của mô hình học trên dữ liệu ẩn danh
anonymized_validation_metrics_s1 = (
    utility_evaluator.evaluate_classification_metrics(

        # Nhãn thật của cùng tập Validation
        y_true=y_anonymized_validation_s1,

        # Nhãn dự đoán từ dữ liệu ẩn danh
        y_pred=y_pred_anonymized_validation_s1,

        # Xác suất dự đoán lớp >50K
        y_prob=y_prob_anonymized_validation_s1,

        # Xác định >50K là lớp dương
        positive_label=">50K"
    )
)


# ============================================================
# 3. HIỂN THỊ METRIC ORIGINAL
# ============================================================


# In tiêu đề kết quả Original
print(
    "===== ORIGINAL VALIDATION ====="
)


# Duyệt từng metric của Original
for metric_name, metric_value in original_validation_metrics_s1.items():

    # Chuyển metric sang phần trăm
    # và làm tròn 2 chữ số thập phân
    print(
        metric_name,
        ":",
        round(
            metric_value * 100,
            2
        ),
        "%"
    )


# ============================================================
# 4. HIỂN THỊ METRIC ANONYMIZED
# ============================================================


# In tiêu đề kết quả Anonymized
print(
    "\n===== ANONYMIZED VALIDATION ====="
)


# Duyệt từng metric của Anonymized
for metric_name, metric_value in anonymized_validation_metrics_s1.items():

    # Chuyển metric sang phần trăm
    # và làm tròn 2 chữ số thập phân
    print(
        metric_name,
        ":",
        round(
            metric_value * 100,
            2
        ),
        "%"
    )
```

```python
# ============================================================
# TÍNH UTILITY RETENTION VÀ UTILITY SCORE
# CHO STRATEGY 1 TRÊN TẬP VALIDATION
# ============================================================


# Gọi hàm calculate_utility_retention()
# để so sánh hiệu quả mô hình:
# Original so với Anonymized
utility_result_validation_s1 = (
    utility_evaluator.calculate_utility_retention(

        # Các metric của mô hình học trên dữ liệu gốc
        original_metrics=original_validation_metrics_s1,

        # Các metric của mô hình học trên dữ liệu ẩn danh
        anonymized_metrics=anonymized_validation_metrics_s1
    )
)


# In tiêu đề để kết quả dễ đọc
print(
    "===== UTILITY RETENTION - STRATEGY 1 ====="
)


# Duyệt từng kết quả Utility Retention
for metric_name, metric_value in utility_result_validation_s1.items():

    # Hiển thị tên chỉ số và giá trị
    # Hàm utility_evaluator đã trả kết quả theo dạng phần trăm
    print(
        metric_name,
        ":",
        round(
            metric_value,
            2
        ),
        "%"
    )
```

```python
# ============================================================
# TEST HÀM 9:
# TỰ ĐỘNG ĐÁNH GIÁ UTILITY CỦA STRATEGY 2
# ============================================================


# Gọi Hàm 9 để tự động đánh giá Machine Learning
# cho dữ liệu ẩn danh của Strategy 2
strategy2_ml_auto_result = (
    ml_evaluator.evaluate_strategy_validation(

        # Dataset gốc ban đầu
        original_dataframe=data_check,

        # Dataset ẩn danh được tạo tự động bởi Strategy 2
        anonymized_dataframe=strategy2_release_auto,

        # Các index gốc còn lại sau suppression của Strategy 2
        retained_indices=strategy2_retained_indices_auto,

        # Cột target của bài toán phân loại
        target_column="thu_nhap",

        # Giữ cùng random_state để có thể tái lập thí nghiệm
        random_state=42
    )
)


# Lấy toàn bộ kết quả Utility Retention
# mà hệ thống vừa tự động tính cho Strategy 2
strategy2_utility_auto = (
    strategy2_ml_auto_result[
        "utility_result"
    ]
)


# In tiêu đề kết quả để dễ theo dõi
print(
    "===== UTILITY AUTO - STRATEGY 2 ====="
)


# Duyệt qua từng chỉ số Utility Retention
for metric_name, metric_value in strategy2_utility_auto.items():

    # Hiển thị tên metric và giá trị
    # làm tròn đến 2 chữ số thập phân
    print(
        metric_name,
        ":",
        round(
            metric_value,
            2
        ),
        "%"
    )


# In riêng Utility Score
# Đây là giá trị sau này dùng để so sánh các Strategy
print(
    "\nUtility Score tự động:",
    round(
        strategy2_ml_auto_result[
            "utility_score_percent"
        ],
        2
    ),
    "%"
)
```

```python
# ============================================================
# KIỂM TRA CHÍNH XÁC THAM SỐ CỦA HÀM rank_strategies()
# ============================================================


# Import inspect để xem cấu trúc
# của hàm trong strategy_selector.py
import inspect


# Hiển thị chữ ký của hàm
# để biết chính xác hàm nhận những tham số nào
print(
    "Chữ ký của rank_strategies:"
)


# In tên hàm và toàn bộ tham số mà hàm chấp nhận
print(
    inspect.signature(
        strategy_selector.rank_strategies
    )
)


# In toàn bộ code của hàm rank_strategies()
# để chúng ta biết chính xác cách truyền trọng số
print(
    "\n===== CODE HÀM rank_strategies ====="
)


# Hiển thị mã nguồn thật đang được notebook sử dụng
print(
    inspect.getsource(
        strategy_selector.rank_strategies
    )
)
```

```python
# ============================================================
# XẾP HẠNG CÁC STRATEGY
# ============================================================


# Import importlib để nạp lại phiên bản mới nhất
# của module strategy_selector
import importlib


# Import module chứa logic tính điểm và xếp hạng Strategy
import strategy_selector


# Nạp lại module để chắc chắn notebook
# đang sử dụng phiên bản mới nhất của strategy_selector.py
strategy_selector = importlib.reload(
    strategy_selector
)


# ============================================================
# GỌI HÀM XẾP HẠNG
# ============================================================


# Gọi hàm rank_strategies()
# để tính Data Retention, Selection Score
# và xếp hạng các Strategy
ranked_strategies = (
    strategy_selector.rank_strategies(

        # Danh sách các Strategy
        # đã được chuẩn bị ở cell trước
        strategies=strategies_for_selection,

        # Trọng số dành cho khả năng giữ lại dữ liệu
        # 0.5 nghĩa là chiếm 50% Selection Score
        weight_data_retention=0.50,

        # Trọng số dành cho Utility Score của Machine Learning
        # 0.5 nghĩa là chiếm 50% Selection Score
        weight_utility=0.50
    )
)


# ============================================================
# HIỂN THỊ KẾT QUẢ XẾP HẠNG
# ============================================================


# Duyệt từng Strategy sau khi đã được xếp hạng
for strategy_result in ranked_strategies:

    # Hiển thị thứ hạng của Strategy
    print(
        "Hạng:",
        strategy_result[
            "rank"
        ]
    )


    # Hiển thị tên Strategy
    print(
        "Strategy:",
        strategy_result[
            "strategy"
        ]
    )


    # Hiển thị tỷ lệ dữ liệu còn được giữ lại
    print(
        "Data Retention (%):",
        round(
            strategy_result[
                "data_retention_%"
            ],
            2
        )
    )


    # Hiển thị Utility Score trên Validation
    print(
        "Utility Score (%):",
        round(
            strategy_result[
                "utility_score_%"
            ],
            2
        )
    )


    # Hiển thị Selection Score
    # Đây là điểm tổng hợp giữa Data Retention và Utility
    print(
        "Selection Score:",
        round(
            strategy_result[
                "selection_score"
            ],
            2
        )
    )


    # Hiển thị Strategy có đạt điều kiện Privacy
    # để được tham gia xếp hạng hay không
    print(
        "Đủ điều kiện?:",
        strategy_result[
            "eligible"
        ]
    )


    # In dòng phân cách để dễ đọc kết quả
    print(
        "----------------------------------------"
    )
```

```python
# ============================================================
# KIỂM TRA HÀM select_best_strategy()
# TRƯỚC KHI GỌI HÀM CHỌN STRATEGY TỐT NHẤT
# ============================================================


# Import inspect để xem chính xác
# hàm select_best_strategy() nhận những tham số nào
import inspect


# Hiển thị chữ ký của hàm
# để tránh truyền sai tên tham số
print(
    "Chữ ký của select_best_strategy:"
)


# In danh sách tham số thật của hàm
print(
    inspect.signature(
        strategy_selector.select_best_strategy
    )
)


# Hiển thị toàn bộ code của hàm
# để kiểm tra nó chọn Strategy tốt nhất theo cách nào
print(
    "\n===== CODE HÀM select_best_strategy ====="
)


# In mã nguồn thật mà notebook đang sử dụng
print(
    inspect.getsource(
        strategy_selector.select_best_strategy
    )
)
```

```python
# ============================================================
# CHỌN STRATEGY TỐT NHẤT
# ============================================================


# Gọi hàm select_best_strategy()
# để lấy Strategy đứng hạng 1
best_strategy = (
    strategy_selector.select_best_strategy(

        # Danh sách các Strategy đã chuẩn bị trước đó
        strategies=strategies_for_selection,

        # Trọng số 50% dành cho khả năng giữ lại dữ liệu
        weight_data_retention=0.50,

        # Trọng số 50% dành cho Utility Score
        weight_utility=0.50
    )
)


# Kiểm tra xem có Strategy nào đạt điều kiện Privacy hay không
if best_strategy is None:

    # Nếu không có Strategy hợp lệ thì thông báo
    print(
        "Không có Strategy nào đạt điều kiện Privacy."
    )

else:

    # Hiển thị Strategy được hệ thống chọn
    print(
        "Strategy tốt nhất:",
        best_strategy[
            "strategy"
        ]
    )


    # Hiển thị thứ hạng
    print(
        "Hạng:",
        best_strategy[
            "rank"
        ]
    )


    # Hiển thị tỷ lệ dữ liệu được giữ lại
    print(
        "Data Retention (%):",
        round(
            best_strategy[
                "data_retention_%"
            ],
            2
        )
    )


    # Hiển thị Utility Score
    print(
        "Utility Score (%):",
        round(
            best_strategy[
                "utility_score_%"
            ],
            2
        )
    )


    # Hiển thị Selection Score
    print(
        "Selection Score:",
        round(
            best_strategy[
                "selection_score"
            ],
            2
        )
    )


    # Hiển thị trạng thái Privacy
    print(
        "Privacy hợp lệ?:",
        best_strategy[
            "privacy_valid"
        ]
    )
```

Strategy 1 được chọn vì vừa đạt yêu cầu Privacy, vừa giữ lại 93.99% dữ liệu, vừa duy trì 97.41% Utility cho Machine Learning, nên được xếp hạng 1 với Selection Score 95.7.

```python
# ============================================================
# LẤY DATASET CỦA STRATEGY TỐT NHẤT
# ============================================================


# Lấy tên Strategy mà hệ thống vừa chọn
# Kết quả hiện tại kỳ vọng là "strategy_1"
selected_strategy_name = (
    best_strategy[
        "strategy"
    ]
)


# Lấy dataset ẩn danh tương ứng
# từ dictionary strategy_results
selected_release_data = (
    strategy_results[
        selected_strategy_name
    ][
        "release_data"
    ]
)


# Lấy danh sách index gốc
# của các bản ghi còn được giữ lại
selected_retained_indices = (
    strategy_results[
        selected_strategy_name
    ][
        "retained_indices"
    ]
)


# Hiển thị tên Strategy được chọn
print(
    "Strategy được chọn:",
    selected_strategy_name
)


# Hiển thị kích thước dataset ẩn danh cuối cùng
# Strategy 1 kỳ vọng có 30.604 dòng và 8 cột
print(
    "Kích thước dataset được chọn:",
    selected_release_data.shape
)


# Hiển thị số lượng index gốc còn giữ lại
# Kỳ vọng bằng đúng 30.604
print(
    "Số retained indices:",
    len(
        selected_retained_indices
    )
)


# Kiểm tra số dòng dataset
# có bằng số lượng retained indices hay không
selected_data_match = (
    len(
        selected_release_data
    )
    ==
    len(
        selected_retained_indices
    )
)


# Hiển thị kết quả kiểm tra
# True nghĩa là dữ liệu đầu ra và index đang khớp nhau
print(
    "Dataset và retained indices khớp nhau?:",
    selected_data_match
)
```

```python
# ============================================================
# LẤY KẾT QUẢ ML VALIDATION
# CỦA STRATEGY ĐƯỢC HỆ THỐNG CHỌN
# ============================================================


# Lấy trực tiếp kết quả ML Validation
# đã được lưu tự động trong strategy_results
selected_ml_result = (
    strategy_results[
        selected_strategy_name
    ][
        "ml_validation"
    ]
)


# ============================================================
# KIỂM TRA
# ============================================================


# Hiển thị tên Strategy được hệ thống lựa chọn
print(
    "Strategy được chọn:",
    selected_strategy_name
)


# Hiển thị Utility Score trên tập Validation
print(
    "Validation Utility Score (%):",
    round(
        selected_ml_result[
            "utility_score_percent"
        ],
        2
    )
)


# Kiểm tra số lượng bản ghi Test
# đã được giữ riêng trong quá trình chia dữ liệu
print(
    "Số bản ghi Test:",
    len(
        selected_ml_result[
            "test_indices"
        ]
    )
)
```

```python
# ============================================================
# KIỂM TRA HÀM FINAL TEST TỰ ĐỘNG
# ============================================================


# Import importlib để nạp lại
# phiên bản mới nhất của ml_evaluator.py
import importlib


# Nạp lại module sau khi vừa thêm hàm mới
ml_evaluator = importlib.reload(
    ml_evaluator
)


# GỌI HÀM ĐÁNH GIÁ TEST TỰ ĐỘNG

# Đánh giá Test cuối cùng của Strategy
# mà hệ thống đã tự động lựa chọn
final_test_auto_result = (
    ml_evaluator.evaluate_selected_strategy_test(

        # Dataset gốc ban đầu
        original_dataframe=data_check,

        # Dataset ẩn danh của Strategy được chọn
        anonymized_dataframe=selected_release_data,

        # Các index còn giữ lại sau Suppression
        retained_indices=selected_retained_indices,

        # Kết quả Validation đã lưu của Strategy được chọn
        validation_result=selected_ml_result,

        # Cột target
        target_column="thu_nhap"
    )
)


# KIỂM TRA KẾT QUẢ


# Hiển thị số bản ghi Test
print(
    "Test size:",
    final_test_auto_result[
        "test_size"
    ]
)


# Kiểm tra hai tập Test có cùng index
print(
    "Index Test giống nhau?:",
    final_test_auto_result[
        "same_test_indices"
    ]
)


# Kiểm tra hai tập Test có cùng target
print(
    "Target Test giống nhau?:",
    final_test_auto_result[
        "same_test_target"
    ]
)


# Hiển thị Accuracy của Original Test
print(
    "Original Accuracy (%):",
    round(
        final_test_auto_result[
            "original_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Accuracy của Anonymized Test
print(
    "Anonymized Accuracy (%):",
    round(
        final_test_auto_result[
            "anonymized_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Utility Score cuối cùng
print(
    "Final Test Utility Score (%):",
    final_test_auto_result[
        "utility_score_percent"
    ]
)
```

```python
# DỰ ĐOÁN TRÊN TẬP TEST CUỐI CÙNG
# CỦA STRATEGY ĐÃ ĐƯỢC HỆ THỐNG CHỌN


# Lấy mô hình Original đã được huấn luyện
# từ kết quả ML của Strategy được chọn
selected_original_model = (
    selected_ml_result[
        "original_model_pipeline"
    ]
)


# Lấy mô hình Anonymized đã được huấn luyện
# từ kết quả ML của Strategy được chọn
selected_anonymized_model = (
    selected_ml_result[
        "anonymized_model_pipeline"
    ]
)


# 1. DỰ ĐOÁN ORIGINAL TEST


# Dùng mô hình Original dự đoán nhãn
# trên 6.121 bản ghi Test
y_pred_original_test_final = (
    selected_original_model.predict(
        X_original_test_final
    )
)


# Lấy danh sách các lớp
# mà mô hình Original đã học
original_test_classes = (
    selected_original_model
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
# để lấy đúng xác suất cho ROC-AUC
original_test_positive_index = (
    list(
        original_test_classes
    )
    .index(
        ">50K"
    )
)


# Lấy xác suất dự đoán lớp >50K
# của từng bản ghi Original Test
y_prob_original_test_final = (
    selected_original_model.predict_proba(
        X_original_test_final
    )[
        :,
        original_test_positive_index
    ]
)


# 2. DỰ ĐOÁN ANONYMIZED TEST


# Dùng mô hình Anonymized dự đoán nhãn
# trên chính 6.121 bản ghi Test tương ứng
y_pred_anonymized_test_final = (
    selected_anonymized_model.predict(
        X_anonymized_test_final
    )
)


# Lấy danh sách các lớp
# mà mô hình Anonymized đã học
anonymized_test_classes = (
    selected_anonymized_model
    .named_steps["model"]
    .classes_
)


# Tìm vị trí của lớp dương >50K
# trong mô hình Anonymized
anonymized_test_positive_index = (
    list(
        anonymized_test_classes
    )
    .index(
        ">50K"
    )
)


# Lấy xác suất dự đoán lớp >50K
# của từng bản ghi Anonymized Test
y_prob_anonymized_test_final = (
    selected_anonymized_model.predict_proba(
        X_anonymized_test_final
    )[
        :,
        anonymized_test_positive_index
    ]
)


# 3. KIỂM TRA SỐ LƯỢNG KẾT QUẢ DỰ ĐOÁN


# Kiểm tra số nhãn dự đoán Original
print(
    "Số dự đoán Original Test:",
    len(
        y_pred_original_test_final
    )
)


# Kiểm tra số xác suất Original
print(
    "Số xác suất Original Test:",
    len(
        y_prob_original_test_final
    )
)


# Kiểm tra số nhãn dự đoán Anonymized
print(
    "Số dự đoán Anonymized Test:",
    len(
        y_pred_anonymized_test_final
    )
)


# Kiểm tra số xác suất Anonymized
print(
    "Số xác suất Anonymized Test:",
    len(
        y_prob_anonymized_test_final
    )
)
```

```python
# TẠO BẢN TÓM TẮT KẾT QUẢ CUỐI CÙNG CỦA HỆ THỐNG


# Lấy kết quả của Strategy đã được hệ thống lựa chọn
selected_strategy_result = (
    strategy_results[
        selected_strategy_name
    ]
)


# Tạo dictionary tổng hợp các thông tin quan trọng
# của toàn bộ quá trình ẩn danh hóa và đánh giá
final_system_summary = {

    # Strategy được hệ thống tự động lựa chọn
    "selected_strategy": selected_strategy_name,

    # Số bản ghi của dữ liệu gốc ban đầu
    "original_records": len(
        data_check
    ),

    # Số bản ghi còn lại trong dữ liệu phát hành
    "released_records": len(
        selected_release_data
    ),

    # Số bản ghi đã bị suppression
    "suppressed_records": (
        len(data_check)
        -
        len(selected_release_data)
    ),

    # Tỷ lệ suppression của Strategy được chọn
    "suppression_percent": (
        selected_strategy_result[
            "suppression_percent"
        ]
    ),

    # Tỷ lệ dữ liệu được giữ lại
    "data_retention_percent": (
        best_strategy[
            "data_retention_%"
        ]
    ),

    # Strategy có đạt điều kiện Privacy
    # theo tiêu chí k và l của đồ án hay không
    "privacy_valid": (
        best_strategy[
            "privacy_valid"
        ]
    ),

    # Utility Score trên Validation
    # được dùng trong quá trình lựa chọn Strategy
    "validation_utility_score_percent": (
        selected_strategy_result[
            "utility_score_percent"
        ]
    ),

    # Selection Score dùng để xếp hạng Strategy
    "selection_score": (
        best_strategy[
            "selection_score"
        ]
    ),

    # Số bản ghi của tập Test cuối cùng
    "test_size": (
        selected_strategy_result[
            "final_test"
        ][
            "test_size"
        ]
    ),

    # Utility Score cuối cùng trên tập Test
    "final_test_utility_score_percent": (
        selected_strategy_result[
            "final_test"
        ][
            "utility_score_percent"
        ]
    )
}



# HIỂN THỊ BẢN TÓM TẮT

# Duyệt từng thông tin để kiểm tra kết quả
for key, value in final_system_summary.items():

    # Hiển thị tên thông tin và giá trị tương ứng
    print(
        key,
        ":",
        value
    )
```

```python
# BỔ SUNG CHI TIẾT MACHINE LEARNING TEST


# Cập nhật các metric của Original và Anonymized
# vào final_system_summary để sau này lưu báo cáo
final_system_summary.update({

    # --------------------------------------------------------
    # METRIC CỦA DỮ LIỆU GỐC - ORIGINAL TEST
    # --------------------------------------------------------

    # Accuracy của mô hình trên dữ liệu gốc
    "original_test_accuracy_percent": (
        original_test_metrics["accuracy"] * 100
    ),

    # Precision của mô hình trên dữ liệu gốc
    "original_test_precision_percent": (
        original_test_metrics["precision"] * 100
    ),

    # Recall của mô hình trên dữ liệu gốc
    "original_test_recall_percent": (
        original_test_metrics["recall"] * 100
    ),

    # F1 Score của mô hình trên dữ liệu gốc
    "original_test_f1_percent": (
        original_test_metrics["f1_score"] * 100
    ),

    # Balanced Accuracy của mô hình trên dữ liệu gốc
    "original_test_balanced_accuracy_percent": (
        original_test_metrics["balanced_accuracy"] * 100
    ),

    # ROC-AUC của mô hình trên dữ liệu gốc
    "original_test_roc_auc_percent": (
        original_test_metrics["roc_auc"] * 100
    ),


    # METRIC CỦA DỮ LIỆU ĐÃ ẨN DANH - ANONYMIZED TEST

    # Accuracy sau ẩn danh hóa
    "anonymized_test_accuracy_percent": (
        anonymized_test_metrics["accuracy"] * 100
    ),

    # Precision sau ẩn danh hóa
    "anonymized_test_precision_percent": (
        anonymized_test_metrics["precision"] * 100
    ),

    # Recall sau ẩn danh hóa
    "anonymized_test_recall_percent": (
        anonymized_test_metrics["recall"] * 100
    ),

    # F1 Score sau ẩn danh hóa
    "anonymized_test_f1_percent": (
        anonymized_test_metrics["f1_score"] * 100
    ),

    # Balanced Accuracy sau ẩn danh hóa
    "anonymized_test_balanced_accuracy_percent": (
        anonymized_test_metrics["balanced_accuracy"] * 100
    ),

    # ROC-AUC sau ẩn danh hóa
    "anonymized_test_roc_auc_percent": (
        anonymized_test_metrics["roc_auc"] * 100
    ),


    # UTILITY RETENTION TRÊN TEST


    # Mức giữ lại Accuracy
    "test_accuracy_retention_percent": (
        final_test_utility["accuracy_retention_%"]
    ),

    # Mức giữ lại F1 Score
    "test_f1_retention_percent": (
        final_test_utility["f1_retention_%"]
    ),

    # Mức giữ lại Balanced Accuracy
    "test_balanced_accuracy_retention_percent": (
        final_test_utility[
            "balanced_accuracy_retention_%"
        ]
    ),

    # Mức giữ lại ROC-AUC
    "test_roc_auc_retention_percent": (
        final_test_utility["roc_auc_retention_%"]
    )
})


# KIỂM TRA CÁC GIÁ TRỊ VỪA BỔ SUNG

print(
    "Original Accuracy (%):",
    round(
        final_system_summary[
            "original_test_accuracy_percent"
        ],
        2
    )
)

print(
    "Original F1 (%):",
    round(
        final_system_summary[
            "original_test_f1_percent"
        ],
        2
    )
)

print(
    "Anonymized Accuracy (%):",
    round(
        final_system_summary[
            "anonymized_test_accuracy_percent"
        ],
        2
    )
)

print(
    "Anonymized F1 (%):",
    round(
        final_system_summary[
            "anonymized_test_f1_percent"
        ],
        2
    )
)

print(
    "Final Test Utility Score (%):",
    final_system_summary[
        "final_test_utility_score_percent"
    ]
)
```

```python
# KIỂM TRA CẤU TRÚC HIỆN TẠI CỦA strategy_results


# Duyệt qua từng Strategy đang có trong hệ thống
for strategy_name in strategy_names:

    # Hiển thị tên Strategy
    print(
        "Strategy:",
        strategy_name
    )


    # Hiển thị toàn bộ tên key
    # đang được lưu trong Strategy hiện tại
    print(
        "Các key:",
        list(
            strategy_results[
                strategy_name
            ].keys()
        )
    )


    # In dòng phân cách để dễ theo dõi
    print(
        "----------------------------------------"
    )
```

```python
# TỰ ĐỘNG ĐÁNH GIÁ ML VALIDATION CHO TẤT CẢ STRATEGY


# Duyệt qua toàn bộ Strategy đang có trong hệ thống
# Ví dụ hiện tại gồm strategy_1 và strategy_2
for strategy_name in strategy_names:

    # 1. LẤY DỮ LIỆU CỦA STRATEGY HIỆN TẠI

    # Lấy dataset đã được ẩn danh hóa
    strategy_release_data = (
        strategy_results[
            strategy_name
        ][
            "release_data"
        ]
    )


    # Lấy index các bản ghi còn được giữ lại
    # sau Generalization + Suppression
    strategy_retained_indices = (
        strategy_results[
            strategy_name
        ][
            "retained_indices"
        ]
    )


    # 2. TỰ ĐỘNG CHẠY MACHINE LEARNING VALIDATION

    # Hàm evaluate_strategy_validation() sẽ:
    # - căn chỉnh dữ liệu Original và Anonymized
    # - chia Train / Validation / Test
    # - huấn luyện Logistic Regression
    # - đánh giá trên Validation
    # - tính Utility Retention
    ml_validation_result = (
        ml_evaluator.evaluate_strategy_validation(

            # Dataset gốc ban đầu
            original_dataframe=data_check,

            # Dataset ẩn danh của Strategy hiện tại
            anonymized_dataframe=strategy_release_data,

            # Index còn được giữ lại
            retained_indices=strategy_retained_indices,

            # Cột cần dự đoán
            target_column="thu_nhap",

            # Giữ cố định random state
            # để kết quả có thể tái lập
            random_state=42
        )
    )


    # 3. LƯU TOÀN BỘ KẾT QUẢ ML VÀO STRATEGY

    # Lưu cả model, metric, split index,
    # Utility Result,... vào chính Strategy đó
    strategy_results[
        strategy_name
    ][
        "ml_validation"
    ] = ml_validation_result


    # Lấy riêng Utility Score
    # để Strategy Selector sử dụng trực tiếp
    strategy_results[
        strategy_name
    ][
        "utility_score_percent"
    ] = (
        ml_validation_result[
            "utility_score_percent"
        ]
    )


    # 4. HIỂN THỊ KẾT QUẢ

    print(
        "Strategy:",
        strategy_name
    )

    print(
        "Validation Utility Score (%):",
        round(
            strategy_results[
                strategy_name
            ][
                "utility_score_percent"
            ],
            2
        )
    )

    print(
        "Đã lưu ml_validation?:",
        "ml_validation"
        in strategy_results[
            strategy_name
        ]
    )

    print(
        "----------------------------------------"
    )
```

```python
# KIỂM TRA HÀM FINAL TEST TỰ ĐỘNG



# Import importlib để nạp lại
# phiên bản mới nhất của ml_evaluator.py
import importlib


# Nạp lại module sau khi vừa thêm hàm mới
ml_evaluator = importlib.reload(
    ml_evaluator
)


# GỌI HÀM ĐÁNH GIÁ TEST TỰ ĐỘNG

# Đánh giá Test cuối cùng của Strategy
# mà hệ thống đã tự động lựa chọn
final_test_auto_result = (
    ml_evaluator.evaluate_selected_strategy_test(

        # Dataset gốc ban đầu
        original_dataframe=data_check,

        # Dataset ẩn danh của Strategy được chọn
        anonymized_dataframe=selected_release_data,

        # Các index còn giữ lại sau Suppression
        retained_indices=selected_retained_indices,

        # Kết quả Validation đã lưu của Strategy được chọn
        validation_result=selected_ml_result,

        # Cột target
        target_column="thu_nhap"
    )
)


# KIỂM TRA KẾT QUẢ


# Hiển thị số bản ghi Test
print(
    "Test size:",
    final_test_auto_result[
        "test_size"
    ]
)


# Kiểm tra hai tập Test có cùng index
print(
    "Index Test giống nhau?:",
    final_test_auto_result[
        "same_test_indices"
    ]
)


# Kiểm tra hai tập Test có cùng target
print(
    "Target Test giống nhau?:",
    final_test_auto_result[
        "same_test_target"
    ]
)


# Hiển thị Accuracy của Original Test
print(
    "Original Accuracy (%):",
    round(
        final_test_auto_result[
            "original_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Accuracy của Anonymized Test
print(
    "Anonymized Accuracy (%):",
    round(
        final_test_auto_result[
            "anonymized_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Utility Score cuối cùng
print(
    "Final Test Utility Score (%):",
    final_test_auto_result[
        "utility_score_percent"
    ]
)
```

```python
# LƯU FINAL TEST TỰ ĐỘNG VÀO STRATEGY ĐƯỢC CHỌN


# Lưu toàn bộ kết quả Final Test tự động
# vào đúng Strategy mà hệ thống đã lựa chọn
strategy_results[
    selected_strategy_name
][
    "final_test"
] = final_test_auto_result


# KIỂM TRA KẾT QUẢ ĐÃ ĐƯỢC LƯU


# Lấy lại kết quả Final Test
# từ chính strategy_results để kiểm tra
selected_final_test_result = (
    strategy_results[
        selected_strategy_name
    ][
        "final_test"
    ]
)


# Hiển thị Strategy được chọn
print(
    "Strategy:",
    selected_strategy_name
)


# Hiển thị số bản ghi Test
print(
    "Test size:",
    selected_final_test_result[
        "test_size"
    ]
)


# Hiển thị Accuracy của dữ liệu gốc
print(
    "Original Accuracy (%):",
    round(
        selected_final_test_result[
            "original_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Accuracy của dữ liệu ẩn danh
print(
    "Anonymized Accuracy (%):",
    round(
        selected_final_test_result[
            "anonymized_metrics"
        ][
            "accuracy"
        ] * 100,
        2
    )
)


# Hiển thị Final Utility Score
print(
    "Final Test Utility Score (%):",
    selected_final_test_result[
        "utility_score_percent"
    ]
)


# Kiểm tra final_test đã tồn tại
# trong Strategy được chọn hay chưa
print(
    "Đã lưu final_test tự động?:",
    "final_test"
    in strategy_results[
        selected_strategy_name
    ]
)
```

```python
# ĐỒNG BỘ KẾT QUẢ FINAL TEST TỰ ĐỘNG
# CHO CÁC PHẦN TỔNG HỢP KẾT QUẢ PHÍA SAU


# Lấy kết quả Final Test đã được lưu tự động
# của Strategy mà hệ thống lựa chọn
selected_final_test_result = (
    strategy_results[
        selected_strategy_name
    ][
        "final_test"
    ]
)


# Lấy metric của dữ liệu Original
# từ kết quả Final Test tự động
original_test_metrics = (
    selected_final_test_result[
        "original_metrics"
    ]
)


# Lấy metric của dữ liệu Anonymized
# từ kết quả Final Test tự động
anonymized_test_metrics = (
    selected_final_test_result[
        "anonymized_metrics"
    ]
)


# Lấy kết quả Utility Retention
# từ Final Test tự động
final_test_utility = (
    selected_final_test_result[
        "utility_result"
    ]
)


# KIỂM TRA

print(
    "Original Accuracy (%):",
    round(
        original_test_metrics[
            "accuracy"
        ] * 100,
        2
    )
)


print(
    "Anonymized Accuracy (%):",
    round(
        anonymized_test_metrics[
            "accuracy"
        ] * 100,
        2
    )
)


print(
    "Final Test Utility Score (%):",
    final_test_utility[
        "utility_score_%"
    ]
)


print(
    "Nguồn kết quả:",
    "strategy_results -> final_test tự động"
)
```

```python
# KIỂM TRA BẢN TỔNG KẾT
# CÓ KHỚP VỚI FINAL TEST TỰ ĐỘNG HAY KHÔNG

# Lấy Utility Score trực tiếp
# từ Final Test tự động
auto_final_utility = (
    strategy_results[
        selected_strategy_name
    ][
        "final_test"
    ][
        "utility_score_percent"
    ]
)


# Lấy Utility Score đang được lưu
# trong bản tổng kết cuối cùng
summary_final_utility = (
    final_system_summary[
        "final_test_utility_score_percent"
    ]
)


# Kiểm tra hai giá trị có giống nhau không
utility_summary_match = (
    auto_final_utility
    ==
    summary_final_utility
)


# Lấy Accuracy Anonymized trực tiếp
# từ Final Test tự động
auto_anonymized_accuracy = (
    strategy_results[
        selected_strategy_name
    ][
        "final_test"
    ][
        "anonymized_metrics"
    ][
        "accuracy"
    ] * 100
)


# Lấy Accuracy Anonymized
# đang lưu trong bản tổng kết
summary_anonymized_accuracy = (
    final_system_summary[
        "anonymized_test_accuracy_percent"
    ]
)


# Kiểm tra hai Accuracy có khớp nhau không
accuracy_summary_match = (
    abs(
        auto_anonymized_accuracy
        -
        summary_anonymized_accuracy
    )
    < 0.000001
)




print(
    "Final Utility tự động (%):",
    auto_final_utility
)

print(
    "Final Utility trong Summary (%):",
    summary_final_utility
)

print(
    "Utility Summary khớp?:",
    utility_summary_match
)

print(
    "Anonymized Accuracy Summary khớp?:",
    accuracy_summary_match
)
```
